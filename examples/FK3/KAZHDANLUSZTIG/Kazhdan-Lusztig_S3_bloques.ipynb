# El álgebra $\mathfrak{D}=\mathcal{D}(\mathcal{FK}_m\# \Bbbk\mathbb{S}_{3})$
Sea $\mathcal{FK}_m$ el álgebra presentada por generadores

$$
\{ x_{(ij)} \mid i<j \in \mathbb{I}_m \}
$$
y relaciones
$$
\begin{align*}
x_{(ij)}^2 &= 0, && i<j \in \mathbb{I}_m, \\
x_{(ij)}x_{(kl)} - x_{(kl)}x_{(ij)} &= 0, && i,j,k,l \in \mathbb{I}_m, \text{ todos distintos}, \\
x_{(jk)}x_{(ik)} - x_{(ij)}x_{(jk)} + x_{(ik)}x_{(ij)} &= 0, && i<j<k \in \mathbb{I}_m, \\
x_{(ik)}x_{(jk)} - x_{(jk)}x_{(ij)} + x_{(ij)}x_{(ik)} &= 0, && i<j<k \in \mathbb{I}_m.
\end{align*}
$$

## El álgebra $\mathcal{FK}_3$


Para $m=3$, el álgebra $\mathcal{FK}_3$ está generada por
$$
x_{(12)},\quad x_{(13)},\quad x_{(23)}
$$
y las relaciones son:
$$
\begin{aligned}
x_{(12)}^2 &= 0, \\
x_{(13)}^2 &= 0, \\
x_{(23)}^2 &= 0, \\
x_{(23)}x_{(13)} - x_{(12)}x_{(23)} + x_{(13)}x_{(12)} &= 0, \\
x_{(13)}x_{(23)} - x_{(23)}x_{(12)} + x_{(12)}x_{(13)} &= 0.
\end{aligned}
$$

## Relaciones en el álgebra $\mathcal{FK}_3$

### *Nilpotencia cuadrática*:
$$
\begin{align*}
x_{(12)}^2 = x_{(13)}^2 = x_{(23)}^2 = 0,
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
x_{12}^2 = x_{13}^2 = x_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
x^{2}_{(ij)} \;=\; x_{(ij)}x_{(ik)} \;+\; x_{(jk)}x_{(ij)} \;+\; x_{(ik)}x_{(jk)}
\;=\;
x_{(ik)}x_{(ij)} \;+\; x_{(ij)}x_{(jk)} \;+\; x_{(jk)}x_{(ik)}
\;=\; 0,
\quad \text{(dada por } B(V)).
$$

$$
\begin{aligned}
x_{(12)}x_{(13)} + x_{(23)}x_{(12)} + x_{(13)}x_{(23)} 
\quad \text{y} \quad 
x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.
\end{aligned}
$$

```gap
###############################################################################
# x_(ij)^2 = 0, x_(ij)x_(jk)+x_(jk)x_(ik)+x_(ik)x_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3      := [ (1,2), (1,3), (2,3) ];;
generators   := gens{[2..4]};;
relationsFK3 := List( generators, s -> s^2);;
i            := 1;;
j            := 2;;
k            := 3;;
Xij          := generators[Position(gensFK3, (i,j))];;
Xik          := generators[Position(gensFK3, (i,k))];;
Xjk          := generators[Position(gensFK3, (j,k))];;
# primera relación: x_(ij)x_(jk) + x_(jk)x_(ik) + x_(ik)x_(ij) = 0
Add(relationsFK3, Xij*Xjk + Xjk*Xik + Xik*Xij );
# segunda relación: x_(jk)x_(ij) + x_(ik)x_(jk) + x_(ij)x_(ik) = 0
Add(relationsFK3, Xjk*Xij + Xik*Xjk + Xij*Xik );
```

```gap
relsFK3 := GP2NPList(relationsFK3);;
PrintNPList( relsFK3 );
#X12^2 
#X13^2 
#X23^2 
#X12X23 + X23X13 + X13X12 
#X23X12 + X13X23 + X12X13 
```

## Relaciones en el álgebra $\mathcal{FK}_3$ dual

### *Nilpotencia cuadrática*:
$$
\begin{align*}
y_{(12)}^2 = y_{(13)}^2 = y_{(23)}^2 = 0.\\
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
y_{12}^2 = y_{13}^2 = y_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
\begin{align*}
    y_{(23)}y_{(13)} + y_{(12)}y_{(23)} + y_{(13)}y_{(12)} &= 0,&&\Rightarrow& y_3y_1 + y_0y_3 + y_1y_0 &= 0, \\
    y_{(13)}x_{(23)} + y_{(23)}y_{(12)} + y_{(12)}y_{(13)} &= 0,&&\Rightarrow& y_1y_3 + y_3y_0 + y_0y_1 &= 0.
\end{align*}
$$

```gap
###############################################################################
# y_(ij)^2 = 0, y_(ij)y_(jk)+y_(jk)y_(ik)+y_(ik)y_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3     := [ (1,2), (1,3), (2,3) ];;
relsFK3dual := [];;
# Paso 1: Relaciones cuadráticas x_(ij)^2 = 0
for k in [1..Length(gensFK3 )] do
    Add(relsFK3dual, [[ [yIndex[k], yIndex[k]] ], [1]]);;
od;
# Paso 2: Relaciones cúbicas de tres índices distintos
triples := [[1,2,3]];;  # único triple distinto para n=3
for t in triples do
    i := t[1];
    j := t[2];
    k := t[3];
    ij := Position(gensFK3, (i,j));
    ik := Position(gensFK3, (i,k));
    jk := Position(gensFK3, (j,k));
    # primera relación: y_(ij)y_(jk) + y_(jk)y_(ik) + y_(ik)y_(ij) = 0
    Add(relsFK3dual,[[[yIndex[ij],yIndex[jk]],[yIndex[jk],yIndex[ik]],[yIndex[ik],yIndex[ij]]],[1,1,1]]);
    # segunda relación: y_(jk)y_(ij) + y_(ik)y_(jk) + y_(ij)y_(ik) = 0
    Add(relsFK3dual,[[[yIndex[jk],yIndex[ij]],[yIndex[ik],yIndex[jk]],[yIndex[ij],yIndex[ik]]],[1,1,1]]);
od;
```

```gap
#############################################################################
##
#F  QGNAG_HomBasis( <Nrep>, <Mrep> )
##
##  Fully algebra-agnostic: takes two lists of square matrices (images of
##  the SAME set of generators, in the SAME order, under two matrix
##  representations N, M of any algebra -- kG, D(G), the Nichols-algebra
##  presentations, whatever) and returns a record
##      rec( dim := <integer>, basis := <list of matrices> )
##  where <dim> = dim Hom(N,M) (as computed by QGNAG_DimHomAModules, kept
##  here only as a cross-check) and <basis> is an explicit list of
##  intertwining matrices Phi, one per independent copy of N found inside M
##  (or vice versa -- see the Remark below on which direction this is).
##
##  No assumption on semisimplicity is needed for this function itself:
##  Hom is always a vector space and its kernel is always computable this
##  way. Semisimplicity is only used later, by the CALLER, to justify that
##  dim Hom(simple,M) is literally the multiplicity of that simple in M
##  (Schur's lemma) -- see eqn:mult in the paper.
##
QGNAG_HomBasis := function( Nrep, Mrep )
    local n, m, C, i, Ci, Im, In, nullsp, UnVec, PhiList, phi, chk, g;

    if not QGNAG_IsValidMatrixRepresentation( Nrep ) then
        Error( "Nrep is not a valid matrix representation" );
    fi;
    if not QGNAG_IsValidMatrixRepresentation( Mrep ) then
        Error( "Mrep is not a valid matrix representation" );
    fi;
    if Length( Nrep ) <> Length( Mrep ) then
        Error( "representations must have the same number of generators" );
    fi;

    n  := NrCols( Nrep[1] );
    m  := NrCols( Mrep[1] );
    Im := IdentityMat( m );
    In := IdentityMat( n );

    # --- Exactly the system of QGNAG_DimHomAModules, but we keep it ---
    C := [];
    for i in [1 .. Length( Nrep )] do
        if IsZero( Nrep[i] ) and IsZero( Mrep[i] ) then
            continue;
        fi;
        Ci := KroneckerProduct( Im, TransposedMat( Nrep[i] ) ) - KroneckerProduct( Mrep[i], In );
        Append( C, Ci );
    od;

    # --- The kernel of C, not just its rank ---
    if Length( C ) = 0 then
        nullsp := IdentityMat( n * m );     # no constraints: all of Hom
    else
        # C acts on column vectors: C * vec(X) = 0.
        # NullspaceMat(A) returns a basis of { v : v*A = 0 } (row vectors),
        # so we solve against TransposedMat(C) to match GAP's convention.
        nullsp := NullspaceMat( TransposedMat( C ) );
    fi;

    # --- Un-vectorize: flat vector of length n*m  -->  n x m matrix X ---
    # (column-major: the j-th block of length n is the j-th column of X,
    # matching Lemma~\ref{lem:vec} of the paper).
    UnVec := function( v, nn, mm )
        local X, j;
        X := [];
        for j in [1 .. mm] do
            Add( X, v{ [ (j-1)*nn+1 .. j*nn ] } );
        od;
        return TransposedMat( X );          # n x m
    end;

    PhiList := List( nullsp, v -> UnVec( v, n, m ) );

    # --- Self-check: every Phi must intertwine the (transposed) actions. ---
    # If this Error ever fires, it means the row/column convention of your
    # Nrep, Mrep does not match this build of C -- swap the two
    # KroneckerProduct terms above (or transpose Nrep/Mrep on input) and
    # re-run; do NOT silently trust the output otherwise.
    for phi in PhiList do
        for g in [1 .. Length( Nrep )] do
            chk := TransposedMat( Nrep[g] ) * phi - phi * TransposedMat( Mrep[g] );
            if not IsZero( chk ) then
                Error( "QGNAG_HomBasis: Phi fails the intertwiner check -- ",
                       "check the vec/unvec convention against your Nrep/Mrep" );
            fi;
        od;
    od;

    return rec( dim := Length( PhiList ), basis := PhiList );
end;
```

```gap
#############################################################################
##
#F  QGNAG_LabeledEmbeddings( <tensorMats>, <SimplesMn> )
##
##  Given the matrices of a (possibly reducible) representation <tensorMats>
##  (e.g. lambda_i \otimes lambda_j), returns a list of records
##      rec( pos := <index into SimplesMn>, dim := <mult>, basis := <list> )
##  one per simple lambda_s with nonzero Hom(tensorMats, lambda_s), i.e.
##  one per weight actually occurring in the decomposition. Each entry's
##  `basis` is the list of embedding matrices Phi (columns = images of a
##  basis of lambda_s inside tensorMats -- see the direction check on
##  QGNAG_HomBasis) coming straight out of Step 1-3 of the previous turn.
##
QGNAG_LabeledEmbeddings := function( tensorMats, SimplesMn )
    local result, s, matsS, hb;
    result := [];
    for s in [1 .. Length( SimplesMn )] do
        matsS := QGNAG_RepresentationMatrices( SimplesMn[s] );
        hb := QGNAG_HomBasis( tensorMats, matsS );
        if hb.dim > 0 then
            Add( result, rec( pos := s, dim := hb.dim, basis := hb.basis ) );
        fi;
    od;
    return result;
end;
```

```gap
#############################################################################
##
#F  QGNAG_TripleEmbeddingsTree( <SimplesMn>, <i>, <j>, <k>, <bracketFirst> )
##
##  Computes composite embeddings of EVERY target weight f into the triple
##  tensor product lambda_i \otimes lambda_j \otimes lambda_k, bracketed
##  either as (i,j) first (bracketFirst = "IJ") or (j,k) first
##  (bracketFirst = "JK"). Returns a list of records
##      rec( f := <index>, through := <intermediate index e or g>,
##           matrix := <composite embedding, (d_i d_j d_k) x d_f> )
##  one per (intermediate, final) pair with nonzero multiplicity in BOTH
##  steps. Multiplicity-free case assumed for simplicity of the record
##  shape (true for D(S3), confirmed empirically); extend `matrix` to a
##  list if you hit multiplicity > 1 elsewhere.
##
QGNAG_TripleEmbeddingsTree := function( SimplesMn, i, j, k, bracketFirst )
    local a, b, c, matsA, matsB, matsC, tensorAB, embAB, entry,
          matsMid, tensorMidC, embMidC, finalEntry, dC_or_dA,
          IdOnThird, composite, out, Psi, Phi;

    out := [];

    if bracketFirst = "IJ" then
        a := i; b := j; c := k;
    else # "JK": group (j,k) first, then tensor with i on the LEFT
        a := j; b := k; c := i;
    fi;

    matsA := QGNAG_RepresentationMatrices( SimplesMn[a] );
    matsB := QGNAG_RepresentationMatrices( SimplesMn[b] );
    matsC := QGNAG_RepresentationMatrices( SimplesMn[c] );

    # Step 1: decompose lambda_a \otimes lambda_b
    tensorAB := QGNAG_TensorProductOfSimples( SimplesMn[a], SimplesMn[b] );
    embAB := QGNAG_LabeledEmbeddings(
                 QGNAG_RepresentationMatrices( tensorAB ), SimplesMn );

    for entry in embAB do
        # entry.pos = e (or g); entry.basis = list of Phi: lambda_e -> AB

        # Step 2: tensor lambda_e (ABSTRACT matrices) with lambda_c
        matsMid := QGNAG_RepresentationMatrices( SimplesMn[entry.pos] );
        tensorMidC := QGNAG_TensorProductOfSimples(
                          SimplesMn[entry.pos], SimplesMn[c] );
        embMidC := QGNAG_LabeledEmbeddings(
                       QGNAG_RepresentationMatrices( tensorMidC ), SimplesMn );

        for finalEntry in embMidC do
            # finalEntry.pos = f; finalEntry.basis = list of Psi: lambda_f -> e⊗c

            for Phi in entry.basis do
                for Psi in finalEntry.basis do
                    IdOnThird := IdentityMat( NrCols( matsC[1] ) );
                    # composite = ( I_{d_c} ⊗ Phi) * Psi
                    #composite := KroneckerProduct( Phi, IdOnThird ) * Psi;
                    composite := KroneckerProduct( IdOnThird, Phi ) * Psi;   # antes: KroneckerProduct(Phi, IdOnThird)

                    Add( out, rec( f := finalEntry.pos,
                                   through := entry.pos,
                                   matrix := composite,
                                   bracket := bracketFirst ) );
                od;
            od;
        od;
    od;

    return out;
end;
```

```gap
#############################################################################
##
#F  QGNAG_ReorderTreeB_to_TreeA( <vec>, <di>, <dj>, <dk> )
##
##  Derived directly from the confirmed convention of
##  QGNAG_TensorProductOfSimples(X,Y) = Kron(Y_images, X_images) (Y outer).
##  Tree A nests as k-outer,(j-outer,i-inner): posA = k*di*dj + j*di + i.
##  Tree B nests as i-outer,(k-outer,j-inner): posB = i*dj*dk + k*dj + j.
##  Pure index bookkeeping -- no Hom-space search, no gauge ambiguity.
##
QGNAG_ReorderTreeB_to_TreeA := function( vec, di, dj, dk )
    local out, i, j, k, posA, posB;
    out := ListWithIdenticalEntries( di*dj*dk, 0 );
    for i in [0..di-1] do
        for j in [0..dj-1] do
            for k in [0..dk-1] do
                posA := k*di*dj + j*di + i + 1;
                posB := i*dj*dk + k*dj + j + 1;
                out[posA] := vec[posB];
            od;
        od;
    od;
    return out;
end;

#############################################################################
##
#F  QGNAG_FMatrix( <SimplesMn>, <i>, <j>, <k>, <f> )
##
##  Computes the recoupling matrix F^{ijk}_f relating the tree-A basis
##  ((i,j) first) to the tree-B basis ((j,k) first) of the lambda_f-
##  isotypic subspace of lambda_i \otimes lambda_j \otimes lambda_k.
##  Solves the FULL linear system -- no orthonormality assumed, per the
##  warning from the previous turn.
##
QGNAG_FMatrix := function( SimplesMn, i, j, k, f )
    local di, dj, dk, treeA, treeB, colsA, colsB, labelsA, labelsB,
          entry, vcols, c, matB, matBrows, Fcols, col, sol, Fmat;

    di := NrCols( QGNAG_RepresentationMatrices( SimplesMn[i] )[1] );
    dj := NrCols( QGNAG_RepresentationMatrices( SimplesMn[j] )[1] );
    dk := NrCols( QGNAG_RepresentationMatrices( SimplesMn[k] )[1] );

    treeA := Filtered( QGNAG_TripleEmbeddingsTree( SimplesMn, i, j, k, "IJ" ), r -> r.f = f );
    treeB := Filtered( QGNAG_TripleEmbeddingsTree( SimplesMn, i, j, k, "JK" ), r -> r.f = f );

    colsA := []; labelsA := [];
    for entry in treeA do
        for c in [1 .. NrCols( entry.matrix )] do
            Add( colsA, List( [1..NrRows(entry.matrix)], r -> entry.matrix[r][c] ) );
            Add( labelsA, entry.through );
        od;
    od;

    colsB := []; 
    labelsB := [];
    for entry in treeB do
        for c in [1 .. NrCols( entry.matrix )] do
            vcols := List( [1..NrRows(entry.matrix)], r -> entry.matrix[r][c] );
            # SIN reorder -- las bases de Tree A y Tree B ya viven en el MISMO
            # espacio con la MISMA indexación, por la asociatividad posicional
            # de QGNAG_TensorProductOfSimples confirmada arriba.
            Add( colsB, vcols );
            Add( labelsB, entry.through );
        od;
    od;

    if Length( colsA ) <> Length( colsB ) then
        Error( "QGNAG_FMatrix: dimension mismatch between tree A and tree B" );
    fi;

    # matBrows: the basis vectors of B as ROWS, for SolutionMat's convention
    # (SolutionMat(mat, v) finds x such that v = x * mat, x expressing v as
    # a combination of the ROWS of mat).
    matBrows := colsB;   # each colsB[c] is already a row-vector-shaped list

    # Solve, one target vector at a time, column by column of A.
    Fcols := [];
    for col in colsA do
        sol := SolutionMat( matBrows, col );
        if sol = fail then
            Error( "QGNAG_FMatrix: target vector not in the span of tree-B ",
                   "basis -- check the reordering or an upstream embedding error" );
        fi;
        Add( Fcols, sol );
    od;

    # Fcols[c] is the c-th column of F, written as a row; transpose to get
    # F itself with columns indexed by tree-A, rows indexed by tree-B.
    Fmat := TransposedMat( Fcols );

    return rec( F := Fmat, labelsA := labelsA, labelsB := labelsB );
end;
 

```

```gap
#############################################################################
##
#F  QGNAG_TreeConnectingMap( <matsTreeA>, <matsTreeB> )
##
##  Instead of hand-deriving a permutation formula (fragile: depends on the
##  exact Kronecker argument order used internally by
##  QGNAG_TensorProductOfSimples, which reverses its two arguments -- see
##  diagnosis above), find the connecting isomorphism directly via
##  QGNAG_HomBasis. Since tree A and tree B realize the SAME abstract
##  representation lambda_i (x) lambda_j (x) lambda_k (associativity of
##  Rep(D(G))), Hom(treeA, treeB) is guaranteed nonzero; when the
##  representation is multiplicity-free in its own right this Hom space is
##  1-dimensional and gives, up to a global scalar, THE connecting map.
##
QGNAG_TreeConnectingMap := function( matsTreeA, matsTreeB )
    local hb;
    hb := QGNAG_HomBasis( matsTreeA, matsTreeB );
    if hb.dim = 0 then
        Error( "QGNAG_TreeConnectingMap: trees are NOT isomorphic as ",
               "representations -- this should never happen by ",
               "associativity; check matsTreeA/matsTreeB construction" );
    elif hb.dim > 1 then
        Print( "Warning: dim Hom(treeA,treeB) = ", hb.dim,
               " > 1 -- the two trees share more than one isomorphism; ",
               "pick the one compatible with the EXPLICIT composite ",
               "embeddings (matching Phi/Psi labels) rather than an ",
               "arbitrary basis vector of this space.\n" );
    fi;
    return hb.basis;  # list of candidate connecting matrices
end;
```

## 3. Blocks (Section 4.4)

`QGNAG_ClassifySimplesIntoBlocksByRelation` implements the construction above. `FK3_BlocksFromRelation` is a local copy that can use either *or* or *and*;
it is used to check that the definitions give the same partition (the review of the paper points out that (6) uses *or* while §4.4 uses *and*).

```gap
# R_p = { q : t_{p,q}(1) <> 0  (or / and)  t_{q,p}(1) <> 0 };  B_1 = R_p,  B_{k+1} = union of R_q over q in B_k.
FK3_BlocksFromRelation := function( pols, requireBoth )
    local n, R, p, q, tpq, tqp, related, visited, blocks, index, block, blockNext, stable;
    n := Length( pols );
    R := List( [ 1 .. n ], x -> [] );
    for p in [ 1 .. n ] do
        for q in [ 1 .. n ] do
            tpq := Sum( pols[p].(String( q )) );     # value at 1 = sum of the coefficients
            tqp := Sum( pols[q].(String( p )) );
            if requireBoth then
                related := ( tpq <> 0 and tqp <> 0 );
            else
                related := ( tpq <> 0 or tqp <> 0 );
            fi;
            if related then
                Add( R[p], q );
            fi;
        od;
        if not p in R[p] then
            Error( "FK3_BlocksFromRelation: p = ", p, " is not in R_p (t_{p,p}(1) = 0)" );
        fi;
    od;
    visited := List( [ 1 .. n ], x -> false );
    blocks  := rec();
    index   := 1;
    for p in [ 1 .. n ] do
        if not visited[p] then
            block := Set( R[p] );
            repeat
                blockNext := Set( Concatenation( List( block, x -> R[x] ) ) );
                stable    := ( blockNext = block );
                block     := blockNext;
            until stable;
            for q in block do
                visited[q] := true;
            od;
            blocks.(String( index )) := block;
            index := index + 1;
        fi;
    od;
    return blocks;
end;;

if IsBoundGlobal( "QGNAG_ClassifySimplesIntoBlocksByRelation" ) then
    block_classification := QGNAG_ClassifySimplesIntoBlocksByRelation( data_quantum_pols );;
else
    block_classification := FK3_BlocksFromRelation( data_quantum_pols, false );;
fi;

fk3_nblocks := Length( RecNames( block_classification ) );;
for fk3_k in [ 1 .. fk3_nblocks ] do
    fk3_b := block_classification.(String( fk3_k ));
    Print( "B_", fk3_k, " = ", fk3_b, "   ", List( fk3_b, i -> SimpleNames[i] ), "\n" );
od;

fk3_sets := Set( List( [ 1 .. fk3_nblocks ], k -> block_classification.(String( k )) ) );;
fk3_or   := FK3_BlocksFromRelation( data_quantum_pols, false );;
fk3_and  := FK3_BlocksFromRelation( data_quantum_pols, true );;

FK3_Check( "5 blocks", fk3_nblocks = 5 );
FK3_Check( "block sizes: four of size 1 and one of size 4", Collected( List( fk3_sets, Length ) ) = [ [ 1, 4 ], [ 4, 1 ] ] );
FK3_Check( "blocks = {1,3,5,6}, {2}, {4}, {7}, {8}", fk3_sets = Set( [ [ 1, 3, 5, 6 ], [ 2 ], [ 4 ], [ 7 ], [ 8 ] ] ) );
FK3_Check( "the blocks of size 1 are the simple Verma modules (2,4,7,8)",
           Concatenation( Filtered( fk3_sets, b -> Length( b ) = 1 ) ) = [ 2, 4, 7, 8 ] );
FK3_Check( "the classification of the package = the one with 'or'", block_classification = fk3_or );
FK3_Check( "'or' and 'and' give the same blocks", fk3_or = fk3_and );
```

## 4. Coefficient matrix $M$ ($\epsilon=M\nu$), its blocks and their inverses

`QGNAG_BlockDecomposition` splits the matrix of quantum polynomials into diagonal blocks; they must be the blocks of the classification above.
`QGNAG_InvertBlocks` inverts each block over $\mathbb Q(q)$ and stores the determinant (`b.det`).

**What to expect for $\mathcal{FK}_3$** (computed independently, with $x=q^{-1}$): the block $\{1,3,5,6\}$ has determinant $(1-q^{-2})^{2}(1-q^{-6})$;
its inverse is $\mathfrak d^{2}\,N$ with $\mathfrak d=(1-q^{-6})^{-1}$ and $N$ a matrix of polynomials in $q^{-1}$ (the compressed form of the tetrahedron applies with $k=2$,
because $(1-q^{-6})/(1-q^{-2})=1+q^{-2}+q^{-4}$); the smallest common denominator is $(1-q^{-2})(1-q^{-6})$. The four blocks of size 1 are $(1)$.
`block_inverses_FK3.tex` shows the inverse as fractions in $q$; `seriesKL_FK3.tex` should show the compressed form with $\mathfrak d^{2}$ (file `fk3_inversion.tex`, delivered separately, is an independent computation of the same formulas to compare with).
The check that matters is `inverse * matrix = identity`; `QGNAG_TestAdjugateBlocks` is informative only.

```gap
# Functions of the package used in sections 4 and 5.
fk3_needed_B := [ "QGNAG_QuamtumCoefficientsShifted", "QGNAG_BaseIndexOfColumns", "QGNAG_CoefficientMatrix",
                  "QGNAG_BlockDecomposition", "QGNAG_InvertBlocks", "QGNAG_TestCoefficientMatrix", "QGNAG_TestInverseBlocks",
                  "QGNAG_LoadBlockMatricesToLaTeX", "QGNAG_LoadBlockMatrixFormToLaTeX", "QGNAG_LoadBlockInversesToLaTeX",
                  "QGNAG_LoadBlockInverseMatrixFormToLaTeX", "QGNAG_LoadBlockAdjugateFormToLaTeX",
                  "QGNAG_WritePolynomialsKLTeX", "QGNAG_WriteSeriesKLTeX",
                  "QGNAG_LoadQuantumCharactersToLaTeX", "QGNAG_ThePolynomialCoefficients" ];;
fk3_missing_B := Filtered( fk3_needed_B, n -> not IsBoundGlobal( n ) );;
if fk3_missing_B <> [] then
    Print( "[MISSING]  not defined: ", fk3_missing_B, "\n" );
else
    Print( "[ OK ]  all the functions of sections 4 and 5 are available\n" );
fi;

nu_coeffs_pkg := QGNAG_QuamtumCoefficientsShifted( data_record_data_decomp );;     # (sic) name used by the package
coeff_base    := QGNAG_BaseIndexOfColumns( nu_coeffs_pkg );;
coeff_M       := QGNAG_CoefficientMatrix( epsilon_and_coeffs, nu_coeffs_pkg, coeff_base );;
blocks_M      := QGNAG_BlockDecomposition( coeff_M );;
inv_blocks    := QGNAG_InvertBlocks( blocks_M );;

FK3_Check( "coefficient matrix reproduces the Verma characters",
           QGNAG_TestCoefficientMatrix( epsilon_and_coeffs, nu_coeffs_pkg, coeff_base, coeff_M ) = [] );
FK3_Check( "every block is square (rows = columns)", ForAll( blocks_M, b -> b.rows = b.cols ) );
FK3_Check( "blocks of M: sizes 4 x 1, 1 x 4",        Collected( List( blocks_M, b -> Length( b.rows ) ) ) = [ [ 1, 4 ], [ 4, 1 ] ] );
FK3_Check( "classification = block decomposition of M",
           Set( List( [ 1 .. fk3_nblocks ], k -> block_classification.(String( k )) ) ) = Set( List( blocks_M, b -> Set( b.cols ) ) ) );
FK3_Check( "QGNAG_TestInverseBlocks",                QGNAG_TestInverseBlocks( inv_blocks ) = true );
FK3_Check( "inverse * matrix = identity in every block",
           ForAll( inv_blocks, b -> b.inverse * b.matrix = IdentityMat( Length( b.rows ) ) * One( b.matrix[1][1] ) ) );

# Determinants (compare with the expected (1-q^-2)^2 (1-q^-6) of the size-4 block).
for fk3_i in [ 1 .. Length( inv_blocks ) ] do
    if IsBound( inv_blocks[fk3_i].det ) then
        Print( "block ", fk3_i, " ", inv_blocks[fk3_i].cols, ":  det = ", inv_blocks[fk3_i].det, "\n" );
    fi;
od;

# Informative only (not counted in the summary).
fk3_adj := FK3_Try( QGNAG_TestAdjugateBlocks, [ inv_blocks ] );;
if fk3_adj[1] = true and Length( fk3_adj ) > 1 then
    Print( "[INFO]  QGNAG_TestAdjugateBlocks = ", fk3_adj[2], "   (informative)\n" );
else
    Print( "[INFO]  QGNAG_TestAdjugateBlocks raised an error   (informative)\n" );
fi;
```

```gap
QGNAG_LoadAllTPolynomialsToLaTeX := function( filename, data_record_data_decomp, epsilon_and_coeffs, SimpleNames )
    local
        out,
        NuVectorNames,
        epsilon,
        coefficients,
        T,
        p,
        j,
        d,
        k,
        name,
        brace,
        comma,
        pName,
        dName,
        coeff,
        terms,
        term,
        firstTerm,
        theta,
        numShifts,
        shiftCounts,
        simpleName,
        shiftData,
        closebrace;

    out        := OutputTextFile(filename, false);
    # ============================================================
    # The admissible shifted vectors
    # ============================================================
    NuVectorNames := QGNAG_NuVectorNames(data_record_data_decomp);
    shiftData     := []; 
    theta         := Length(epsilon_and_coeffs);    # Number of simples
    for p in [1..theta] do     # Number of admissible shifts for each p.
        shiftData[p] := QGNAG_AllShiftedVectorsLeft( [ data_record_data_decomp[p] ] );
    od;
    shiftCounts := List( shiftData, Length );
    numShifts   := Length(epsilon_and_coeffs[1].coefficients);
    Print("Length(nu) = ", numShifts, "\n"); # Number of shifts for each simple.
    Print("Shift counts = ", shiftCounts, "\n");
    # ============================================================
    # T[p][j][d+1] = f_{p,d}^j
    #
    # p = simple appearing in the shifted vectors
    # j = Verma character epsilon_j
    # d = shift
    # ============================================================
    T := [];
    for p in [1..theta] do
        T[p] := [];
        for j in [1..theta] do
            T[p][j] := [];
            # We only need the shifts which actually occur
            # for the simple p.
            for d in [0..shiftCounts[p]-1] do
                T[p][j][d+1] := 0;
            od;
        od;
    od;
    # ============================================================
    # Compute the coefficients for every epsilon_j
    # ============================================================
    for j in [1..theta] do
        epsilon      := epsilon_and_coeffs[j].vector;
        coefficients := epsilon_and_coeffs[j].coefficients;
        if Length(coefficients) <> numShifts then
            Error( "Unexpected number of coefficients: ", Length(coefficients),  " instead of ", numShifts );
        fi;
        # ----------------------------------------------------
        # Each coefficient corresponds to exactly one
        # shifted vector nu_{p,d}.
        # ----------------------------------------------------
        for k in [1..numShifts] do
            coeff := coefficients[k];
            if coeff <> 0 then
                name := NuVectorNames[k];
                # Expected format:
                #
                #     \nu_{p,d}
                #
                # Extract p and d.
                brace := Position(name, '{');
                comma := Position(name, ',');
                pName := Int( name{ [brace+1..comma-1] } );
                dName := Int( name{ [comma+1..Length(name)-1] } );
                T[pName][j][dName+1] := T[pName][j][dName+1] + coeff;
            fi;
        od;
    od;
    AppendTo( out, "\\section{Polynomials $t_{j,p}(\\boldsymbol{q})$}\n\n" ); # Section
    AppendTo( out, "\\[\nt_{j,p}(\\boldsymbol{q}) = \\sum_{d} f_{p,d}^{j}\\boldsymbol{q}^{d}\n\\]\n\n"); # Formula
    # ============================================================
    # Table
    #
    # Rows: j = 1,...,8
    # Columns: p = 1,...,8
    #
    # Entry:
    #
    #     t_{j,p}(q)
    #
    # ============================================================
    AppendTo( out, "\\[\n" );
    AppendTo( out, "\\begin{array}{c|c" );
    for p in [1..theta] do
        AppendTo(out, "|c");
    od;
    AppendTo( out, "}\n" );
    # Header
    AppendTo( out, "j & \\text{Simple}" );
    for p in [1..theta] do
        AppendTo( out, " & t_{j,", String(p), "}(\\boldsymbol{q})" );
    od;
    AppendTo( out, " \\\\\\hline\n");
    # ============================================================
    # Rows
    # ============================================================
    for j in [1..theta] do
        AppendTo( out, String(j), " & ", SimpleNames[j] );
        for p in [1..theta] do
            terms := [];
            firstTerm := true;
            for d in [0..shiftCounts[p]-1] do
                coeff := T[p][j][d+1];
                if coeff <> 0 then
                    # ------------------------------------------------
                    # Construct q^d
                    # ------------------------------------------------
                    if d = 0 then
                        if coeff = 1 then
                            term := "1";
                        elif coeff = -1 then
                            term := "-1";
                        else
                            term := String(coeff);
                        fi;
                    elif d = 1 then
                        if coeff = 1 then
                            term := "\\boldsymbol{q}";
                        elif coeff = -1 then
                            term := "-\\boldsymbol{q}";
                        else
                            term := Concatenation( String(coeff), "\\,\\boldsymbol{q}" );
                        fi;
                    else
                        if coeff = 1 then
                            term := Concatenation( "\\boldsymbol{q}^{", String(d), "}" );
                        elif coeff = -1 then
                            term := Concatenation( "-\\boldsymbol{q}^{", String(d), "}" );
                        else
                            term := Concatenation( String(coeff), "\\,\\boldsymbol{q}^{", String(d), "}" );
                        fi;
                    fi;
                    # ------------------------------------------------
                    # Add sign between terms
                    # ------------------------------------------------
                    if not firstTerm then
                        if coeff > 0 then
                            term := Concatenation( "+ ", term );
                        fi;
                    fi;
                    Add(terms, term);
                    firstTerm := false;
                fi;
            od;
            if Length(terms) = 0 then
                AppendTo(out, " & 0");
            else
                AppendTo( out, " & ", JoinStringsWithSeparator(terms, " ") );
            fi;
        od;
        AppendTo( out, " \\\\\n");
    od;
    AppendTo( out, "\\end{array}\n");
    AppendTo( out, "\\]\n" );
    CloseStream(out);
end;
```
