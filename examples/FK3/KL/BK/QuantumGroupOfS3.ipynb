# El álgebra $\mathfrak{D}=\mathcal{D}(\mathcal{FK}_m\# \Bbbk\mathbb{S}_{3})$
Sea $\mathcal{FK}_m$ el álgebra presentada por generadores

$$
\{ x_{(ij)} \mid i<j \in \mathbb{I}_m \}
$$
y relaciones
$$
\begin{align*}
x_{(ij)}^2 &= 0, && i<j \in \mathbb{I}_m, \\
x_{(ij)}x_{(kl)} - x_{(kl)}x_{(ij)} &= 0, && i,j,k,l \in \mathbb{I}_m, \text{ todos distintos}, \\
x_{(jk)}x_{(ik)} - x_{(ij)}x_{(jk)} + x_{(ik)}x_{(ij)} &= 0, && i<j<k \in \mathbb{I}_m, \\
x_{(ik)}x_{(jk)} - x_{(jk)}x_{(ij)} + x_{(ij)}x_{(ik)} &= 0, && i<j<k \in \mathbb{I}_m.
\end{align*}
$$

## El álgebra $\mathcal{FK}_3$


Para $m=3$, el álgebra $\mathcal{FK}_3$ está generada por
$$
x_{(12)},\quad x_{(13)},\quad x_{(23)}
$$
y las relaciones son:
$$
\begin{aligned}
x_{(12)}^2 &= 0, \\
x_{(13)}^2 &= 0, \\
x_{(23)}^2 &= 0, \\
x_{(23)}x_{(13)} - x_{(12)}x_{(23)} + x_{(13)}x_{(12)} &= 0, \\
x_{(13)}x_{(23)} - x_{(23)}x_{(12)} + x_{(12)}x_{(13)} &= 0.
\end{aligned}
$$

## Relaciones en el álgebra $\mathcal{FK}_3$

### *Nilpotencia cuadrática*:
$$
\begin{align*}
x_{(12)}^2 = x_{(13)}^2 = x_{(23)}^2 = 0,
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
x_{12}^2 = x_{13}^2 = x_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
x^{2}_{(ij)} \;=\; x_{(ij)}x_{(ik)} \;+\; x_{(jk)}x_{(ij)} \;+\; x_{(ik)}x_{(jk)}
\;=\;
x_{(ik)}x_{(ij)} \;+\; x_{(ij)}x_{(jk)} \;+\; x_{(jk)}x_{(ik)}
\;=\; 0,
\quad \text{(dada por } B(V)).
$$

$$
\begin{aligned}
x_{(12)}x_{(13)} + x_{(23)}x_{(12)} + x_{(13)}x_{(23)} 
\quad \text{y} \quad 
x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.
\end{aligned}
$$

````gap
###############################################################################
# x_(ij)^2 = 0, x_(ij)x_(jk)+x_(jk)x_(ik)+x_(ik)x_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3      := [ (1,2), (1,3), (2,3) ];;
generators   := gens{[2..4]};;
relationsFK3 := List( generators, s -> s^2);;
i            := 1;;
j            := 2;;
k            := 3;;
Xij          := generators[Position(gensFK3, (i,j))];;
Xik          := generators[Position(gensFK3, (i,k))];;
Xjk          := generators[Position(gensFK3, (j,k))];;
# primera relación: x_(ij)x_(jk) + x_(jk)x_(ik) + x_(ik)x_(ij) = 0
Add(relationsFK3, Xij*Xjk + Xjk*Xik + Xik*Xij );
# segunda relación: x_(jk)x_(ij) + x_(ik)x_(jk) + x_(ij)x_(ik) = 0
Add(relationsFK3, Xjk*Xij + Xik*Xjk + Xij*Xik );
````

````gap
relsFK3 := GP2NPList(relationsFK3);;
PrintNPList( relsFK3 );
#X12^2 
#X13^2 
#X23^2 
#X12X23 + X23X13 + X13X12 
#X23X12 + X13X23 + X12X13 
````

## Relaciones en el álgebra $\mathcal{FK}_3$ dual

### *Nilpotencia cuadrática*:
$$
\begin{align*}
y_{(12)}^2 = y_{(13)}^2 = y_{(23)}^2 = 0.\\
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
y_{12}^2 = y_{13}^2 = y_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
\begin{align*}
    y_{(23)}y_{(13)} + y_{(12)}y_{(23)} + y_{(13)}y_{(12)} &= 0,&&\Rightarrow& y_3y_1 + y_0y_3 + y_1y_0 &= 0, \\
    y_{(13)}x_{(23)} + y_{(23)}y_{(12)} + y_{(12)}y_{(13)} &= 0,&&\Rightarrow& y_1y_3 + y_3y_0 + y_0y_1 &= 0.
\end{align*}
$$

````gap
###############################################################################
# y_(ij)^2 = 0, y_(ij)y_(jk)+y_(jk)y_(ik)+y_(ik)y_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3     := [ (1,2), (1,3), (2,3) ];;
relsFK3dual := [];;
# Paso 1: Relaciones cuadráticas x_(ij)^2 = 0
for k in [1..Length(gensFK3 )] do
    Add(relsFK3dual, [[ [yIndex[k], yIndex[k]] ], [1]]);;
od;
# Paso 2: Relaciones cúbicas de tres índices distintos
triples := [[1,2,3]];;  # único triple distinto para n=3
for t in triples do
    i := t[1];
    j := t[2];
    k := t[3];
    ij := Position(gensFK3, (i,j));
    ik := Position(gensFK3, (i,k));
    jk := Position(gensFK3, (j,k));
    # primera relación: y_(ij)y_(jk) + y_(jk)y_(ik) + y_(ik)y_(ij) = 0
    Add(relsFK3dual,[[[yIndex[ij],yIndex[jk]],[yIndex[jk],yIndex[ik]],[yIndex[ik],yIndex[ij]]],[1,1,1]]);
    # segunda relación: y_(jk)y_(ij) + y_(ik)y_(jk) + y_(ij)y_(ik) = 0
    Add(relsFK3dual,[[[yIndex[jk],yIndex[ij]],[yIndex[ik],yIndex[jk]],[yIndex[ij],yIndex[ik]]],[1,1,1]]);
od;
````

# The simple Verma modules

Let us take the expression
$$
\Omega := -2\tau^2\delta_e + 2\tau\sigma\delta_e + 4\sigma\tau\delta_e - 2\tau\delta_e + 2\sigma\delta_e - 2\sigma\tau - 2
4$$. 
Since $\delta_e$ acts as the identity in all cases, we can formally replace
$$
\delta_e \mapsto I.
$$ 
Then
$$
\Omega = -2\tau^2 + 2\tau\sigma + 4\sigma\tau - 2\tau + 2\sigma - 2\sigma\tau - 2.
$$
Grouping:
$$
\Omega = -2\tau^2 + 2\tau\sigma + 2\sigma\tau - 2\tau + 2\sigma - 2.
$$
We will use:
$$
\sigma = (12), \quad \tau = (123).
$$
In $\mathbb{S}_3$,
$$
\tau^2 = (132),
$$
and furthermore
\begin{align*}
\tau\sigma &= (123)(12) = (13), \\
\sigma\tau &= (12) (123) = (23).
\end{align*}
Therefore,
$$
\Omega = -2(132) + 2(13) + 2(23) - 2(123) + 2(12) - 2e.
$$

## The Verma modules $\mathrm{M}(e, +)$ isn't $\mathfrak{D}$-simple.

- 1) trivial representation on $|e\rangle_{+}$

The trivial representation satisfies
$$
\varepsilon(\text{transposition}) = 1, \quad \varepsilon(\text{3-cycle}) = 1.
$$

Then:
\begin{align*}
\varepsilon(\tau) = 1, \quad \varepsilon(\tau^2) = 1, \\
\varepsilon(\sigma) = 1, \quad \varepsilon(\tau\sigma) = 1, \quad \varepsilon(\sigma\tau) = 1.
\end{align*}
We apply this term by term:
\begin{align*}
-2\tau^2 &\mapsto -2(1) = -2, \\
2\tau\sigma &\mapsto 2(1) = 2, \\
2\sigma\tau &\mapsto 2(1) = 2, \\
-2\tau &\mapsto -2(1) = -2, \\
2\sigma &\mapsto 2(1) = 2, \\
-2e &\mapsto -2.
\end{align*}
Adding up:
$$
-2 + 2 + 2 - 2 + 2 - 2 = 0.
$$
Therefore,
$$
\Omega|e\rangle_{+} = 0 |e\rangle_{-} = 0.
$$

The relation is a direct consequence of Schur's Lemma.

Given the decomposition $M \cong \bigoplus_{j=1}^n S_j^{\oplus t_j}$, applying the left-exact functor $\operatorname{Hom}_A(S_i, -)$ yields an isomorphism of vector spaces over the base field $k$:

$$\operatorname{Hom}_A(S_i, M) \cong \bigoplus_{j=1}^n \operatorname{Hom}_A(S_i, S_j)^{\oplus t_j}$$

Because the modules $\{S_1, \dots, S_n\}$ are irreducible and mutually non-isomorphic, Schur's Lemma dictates that $\operatorname{Hom}_A(S_i, S_j) = 0$ whenever $i \neq j$. Consequently, all cross-terms vanish, leaving:

$$\operatorname{Hom}_A(S_i, M) \cong \operatorname{Hom}_A(S_i, S_i)^{\oplus t_i}$$

By Schur's Lemma, $\operatorname{Hom}_A(S_i, S_i)$ is a division algebra over $k$. Taking dimensions over $k$ gives the explicit formula for the multiplicity $t_i$:

$$t_i = \frac{\dim_k \operatorname{Hom}_A(S_i, M)}{\dim_k \operatorname{Hom}_A(S_i, S_i)}$$

If the base field $k$ is algebraically closed, $\operatorname{Hom}_A(S_i, S_i) \cong k$, making its dimension $1$. In this case, the relation reduces entirely to:

$$t_i = \dim_k \operatorname{Hom}_A(S_i, M)$$

````gap
QGNAG_Coproduct:=function(g, groupElems)
    local relsxBos, h, pos_h, h_index, pos_j, action, chi_j_h,
          pos_action, leftMon, rightMon;

    relsxBos := [];;
    # Iterate over all elements g in the group structure
    for h in groupElems do
        action    := h * ElementFq( j ); # Compute the action of the group element on the field component of j
        pos_action := Position( QGNAG.Config.gensA, action!.FieldPart );
        pos_j     := Position( QGNAG.Config.gensA, j );
        chi_j_h   := ChiForSemidirectProduct( h ); # Compute the $\chi_{j}(g)$ factor for the semidirect product
        pos_h     := Position( QGNAG.Config.elmsG, h );
        h_index   := QGNAG.Config.groupIndex[pos_h];
        #Print("QGNAG.Config.groupIndex: ", QGNAG.Config.groupIndex[pos_g], "\n" );
        if pos_action <> fail then
            leftMon  := Concatenation( h_index, [ QGNAG.Config.xIndex[pos_j] ] );     # Construct the monomials: g * x_j           
            rightMon := Concatenation( [ QGNAG.Config.xIndex[pos_action] ], h_index ); # Construct the transformed monomials: x_{g * j} * g
            Add(relsxBos, [ [ leftMon, rightMon ], [1, -chi_j_h] ] );    # Add the relation in the form: leftMon - chi_j_g * rightMon = 0
        fi;
    od;
    return relsxBos;
end;
````

````gap
ActionMatrixByGenerator := function( BaseNichols, J_Nichols, gen )
    local N, Amat, Lt, genGi, bj, pol, mon, PolG2, PolG2Norm, pos, coeffss, bi;
    N := Length(BaseNichols);
    Amat := NullMat(N, N);
    # Precálculos fuera del bucle
    Lt := List(QGNAG.Config.gensA, ElementFq);
    genGi := QGNAG.Config.elmsG[Position( allPairsInG, gen )];
    for bj in [1..N] do
        pol := BaseNichols[bj];
        mon := pol[1];
        PolG2     := [[List(mon[1], m -> Position(Lt, genGi * QGNAG.Config.gensA[m]))], [1]];
        PolG2Norm := StrongNormalFormNP(PolG2, J_Nichols);
        pos       := List(PolG2Norm[1], tx -> Position(BaseNichols, [[tx], [1]]));
        coeffss   := PolG2Norm[2] * ChiForSemidirectProduct( genGi )^Length(pol[1][1]);
        for bi in pos do
            Amat[bi][bj] := coeffss[Position(pos, bi)];
        od;
    od;
    return Amat;
end;
````

````gap
ActionBlocksByGenerator := function( BaseNichols, J_Nichols, gen )
    local MaxDeg, MatricesPorGrado, Lt, genGi, d, Base_d, dim_d, M_d, j, pol, mon, mon_act, PolGd, PolGdNorm, pos, coeffs, k, row, col;

    # Precálculos fuera de los bucles de grado
    Lt := List(gensA, ElementFq);
    genGi := elmsG[Position( allPairsInG, gen )];

    # Determinar el grado máximo presente en la base
    MaxDeg := Maximum(List(BaseNichols, p -> Length(p[1][1])));
    MatricesPorGrado := rec();

    for d in [0..MaxDeg] do
        # Filtramos solo los elementos de la base con grado tensorial 'd'
        Base_d := Filtered(BaseNichols, p -> Length(p[1][1]) = d);
        dim_d := Length(Base_d);

        if dim_d = 0 then
            continue;
        fi;

        M_d := NullMat(dim_d, dim_d, Rationals);

        for j in [1..dim_d] do
            pol := Base_d[j];
            mon := pol[1];

            mon_act := List(mon[1], m -> Position(Lt, genGi * gensA[m]));
            PolGd := [[mon_act], [1]];
            PolGdNorm := StrongNormalFormNP(PolGd, J_Nichols);

            pos := List(PolGdNorm[1], tx -> Position(Base_d, [[tx], [1]]));
            coeffs := PolGdNorm[2] * ChiForSemidirectProduct( genGi )^d;

            for k in [1..Length(pos)] do
                row := pos[k];
                col := j;
                M_d[row][col] := coeffs[k];
            od;
        od;

        # Guardar el bloque matricial en el registro con la clave del grado
        MatricesPorGrado.(String(d)) := M_d;
    od;

    return MatricesPorGrado;
end;
````

Supongamos que las *reglas de fusión* de $H$ son
$$
M_i\otimes M_s=\bigoplus_{k=1}^\theta M_k^{N^{i s}_k}
$$

aquí calculamos 
$$
{N^{i s}_k}=\dim \operatorname{Hom}_H(M_i\otimes M_s, M_k)
$$ 

Sea $\mathfrak{B}=\bigoplus\limits_{i,j} M_i[j]^{b_{i j}}$ la descomposición

aquí calculamos 
$$
{b_{i j}}=\dim \operatorname{Hom}_H(M_i, \mathfrak{B}^j)

Dado $M_s$, calcular
$$
d_{kj}^s=\sum_{i=1}^{\theta}N_k^{i s}b_{ij}=\sum_{i=1}^{\theta}\dim \operatorname{Hom}_H(M_i\otimes M_s, M_k)\cdot\dim \operatorname{Hom}_H(M_i, \mathfrak{B}^j).
$$

### $\texttt{QGNAG\_GradedNicholsMultiplicity}$

Sea $\mathfrak{B}=\bigoplus_{i,j} M_i[j]^{\,b_{ij}}$
la descomposición de la álgebra de Nichols $\mathfrak{B}$ como
$D(G)$-módulo graduado, donde $M_i$ denota el $i$-ésimo módulo simple de
$D(G)$ y $M_i[j]$ indica que dicho módulo aparece en grado $j$.

La información correspondiente a esta descomposición se almacena en el
record $\texttt{rec\_info\_nichols}$. Para cada grado $j$, el componente
$\texttt{rec\_info\_nichols}.j$ contiene una lista de pares $[b_{ij},i]$
para aquellos índices $i$ tales que $b_{ij}\neq 0$. En otras palabras,
$\texttt{rec\_info\_nichols}.j=\bigl\{[b_{ij},i] : b_{ij}\neq 0\bigr\}.$

Por ejemplo, si $\texttt{rec\_info\_nichols.2 = [[1,43],[1,44]]},$
esto significa que $\mathfrak{B}^2 \simeq M_{43}[2]\oplus M_{44}[2].$
Por lo tanto, $b_{43,2}=1, \qquad b_{44,2}=1,$ mientras que $b_{i,2}=0$
para todo $i\notin\{43,44\}$.

La función
$
\texttt{QGNAG\_GradedNicholsMultiplicity}(i,j,
\texttt{rec\_info\_nichols})
$
permite recuperar directamente la multiplicidad $b_{ij}$ a partir de
esta información.

Su implementación es
```python
InstallGlobalFunction(QGNAG_GradedNicholsMultiplicity, function( i, j, rec_info_nichols )
        local decomposition, pair;

        if not IsBound( rec_info_nichols.(String(j)) ) then
            return 0;
        fi;
        decomposition := rec_info_nichols.(String(j));
        for pair in decomposition do
            if pair[2] = i then
                return pair[1];
            fi;
        od;
        return 0;
    end
);
```

La función primero verifica si el grado $j$ aparece en
$\texttt{rec\_info\_nichols}$. Si no aparece, entonces el correspondiente
componente homogéneo es nulo y se devuelve $b_{ij}=0.$
Si el grado está presente, se recorre la lista
$\texttt{rec\_info\_nichols}.j$ buscando un par cuya segunda componente sea $i$. Si se encuentra el par
$[b_{ij},i],$ la función devuelve su primera componente, es decir, $b_{ij}$.
Si no se encuentra dicho índice, entonces $M_i[j]$ no aparece en la
descomposición y se devuelve $0$.

Por ejemplo, $\texttt{QGNAG\_GradedNicholsMultiplicity}(43,2,\texttt{rec\_info\_nichols})$
devuelve $1,$ pues $[1,43]\in\texttt{rec\_info\_nichols}.2.$
En cambio,$\texttt{QGNAG\_GradedNicholsMultiplicity}(1,2,\texttt{rec\_info\_nichols})$ devuelve $0,$ pues $[b_{1,2},1]$ no aparece en la descomposición de $\mathfrak{B}_2$.

````gap
QGNAG_GradedMultiplicity:=function( i, j, rec_info )
    local decomposition, pair;
    if not IsBound( rec_info.(String(j)) ) then
        return 0;
    fi;
    decomposition := rec_info.(String(j));
    for pair in decomposition do
        if pair[2] = i then
            return pair[1];
        fi;
    od;
    return 0;
end;
````

For a fixed simple module $M_s$, the function constructs the multiplicity table
$$
d_{sk}^{\,j}
=
\sum_{i=1}^{n} N_k^{i,s} b_{ij},
\qquad
0\leq j\leq \ell,\quad 1\leq k\leq n.
$$
The table has $\ell+1$ rows and $n$ columns: each row corresponds to a
degree $j$ of $\mathfrak{B}$, while each column corresponds to a simple
module $M_k$. Thus, the $(j,k)$-entry is the multiplicity of $M_k$ in
$M_s\otimes\mathfrak{B}_j$.

````gap
QGNAG_GradedFusionByDegree:=function( s, SimplesMn, rec_info )
    local theta, data, j, k, r;
    theta := Maximum( List( RecNames(rec_info), Int ) );
    data  := [];
    for k in [1 .. Length(SimplesMn)] do
        r := rec();
        for j in [0 .. theta] do
            r.(String(j)) := QGNAG_GradedFusionMultiplicity( s, k, j, SimplesMn, rec_info );
        od;
        Add(data, r);
    od;
    return data;
end;
````

For a fixed simple module $M_s$, the function vectorizes row-wise the
multiplicity table
$$
\bigl(d_{sk}^{\,j}\bigr)_{\substack{1\leq k\leq \theta\\0\leq j\leq\ell}},
$$
where the rows are indexed by the simple modules $M_k$ and the columns
by the degrees $j$. The vector is constructed degree by degree, and for
each degree $j$ it lists the multiplicities for $k=1,\ldots,\theta$:
$$
\varepsilon_s=
\left(
d_{s1}^{0},\ldots,d_{s \theta}^{0},
d_{s1}^{1},\ldots,d_{s \theta}^{1},
\ldots,
d_{s1}^{\ell},\ldots,d_{s \theta}^{\ell}
\right)
\in\mathbb{Q}^{\theta(\ell+1)}.
$$

````gap
QGNAG_GradedFusionVector := function( s, SimplesMn, rec_info )
    local data, epsilon, j, k, theta;
    data    := QGNAG_GradedFusionByDegree( s, SimplesMn, rec_info );
    theta   := Maximum( List( RecNames(rec_info), Int ) );
    epsilon := [];
    for j in [0 .. theta] do
        for k in [1 .. Length(SimplesMn)] do
            Add(epsilon, data[k].(String(j)));
        od;
    od;
    return epsilon;
end;
````

For a fixed simple module $M_s$, where $s=1,\ldots,\theta$, the data structure
$\texttt{data\_s}$ stores the graded fusion multiplicities
$
\texttt{data\_s}[k].(j)=d_{sk}^{\,j},
\qquad
k=1,\ldots,\theta,\quad j=0,\ldots,\ell.
$
Thus, $\texttt{data\_s}$ is an $(\ell+1)\times \theta$ table whose rows are
indexed by the simple modules $M_k$ and whose columns are indexed by
the degrees $j$.

The function $\texttt{QGNAG\_GradedFusionVectorFromData}$ vectorizes this
precomputed table degree by degree, producing
$$
\varepsilon_s=
\left(
d_{s1}^{0},\ldots,d_{s\theta}^{0},
d_{s1}^{1},\ldots,d_{s\theta}^{1},
\ldots,
d_{s1}^{\ell},\ldots,d_{s\theta}^{\ell}
\right)
\in\mathbb{Q}^{N(\ell+1)}.
$$
Here, $s,k=1,\ldots,N$ index the simple modules, while
$j=0,\ldots,\ell$ indexes the grading of the Nichols algebra.

````gap
QGNAG_GradedFusionVectorFromData := function(data_s, SimplesMn )
    local epsilon, degrees, j, k;
    degrees := List( RecNames(data_s[1]), Int);
    Sort(degrees);
    epsilon := [];
    for j in degrees do
        for k in [1 .. Length(SimplesMn)] do
            Add( epsilon, data_s[k].(String(j)) );
        od;
    od;
    return epsilon;
end;
````

````gap
QGNAG_TestGradedFusionVectorFromData := function( data_s, SimplesMn )
    local expected, actual, degrees, j, k;
    degrees := List( RecNames(data_s[1]), Int );
    Sort(degrees);
    expected := [];
    for j in degrees do
        for k in [1 .. Length(SimplesMn)] do
            Add( expected, data_s[k].(String(j)));
        od;
    od;
    actual   := QGNAG_GradedFusionVectorFromData( data_s, SimplesMn);
    if actual = expected then
        Print("Test passed.\n");
        return true;
    else
        Print("Test failed.\n");
        Print("Expected: ", expected, "\n");
        Print("Actual:   ", actual, "\n");
        return false;
    fi;
end;
````

````gap
QGNAG_TestGradedFusionVectorFromData := function( data_s, SimplesMn )
    local epsilon, degrees, N, j, k, p;
    epsilon := QGNAG_GradedFusionVectorFromData( data_s, SimplesMn );
    degrees := List( RecNames(data_s[1]), Int );
    Sort(degrees);
    N       := Length(SimplesMn);
    for j in degrees do
        for k in [1 .. N] do
            p := j * N + k;
            if epsilon[p] <> data_s[k].(String(j)) then
                Print("Test failed at j = ", j, ", k = ", k, ".\n");
                return false;
            fi;
        od;
    od;
    Print("Test passed.\n");
    return true;
end;
````
