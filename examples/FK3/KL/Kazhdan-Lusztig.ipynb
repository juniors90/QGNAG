# El álgebra $\mathfrak{D}=\mathcal{D}(\mathcal{FK}_m\# \Bbbk\mathbb{S}_{3})$
Sea $\mathcal{FK}_m$ el álgebra presentada por generadores

$$
\{ x_{(ij)} \mid i<j \in \mathbb{I}_m \}
$$
y relaciones
$$
\begin{align*}
x_{(ij)}^2 &= 0, && i<j \in \mathbb{I}_m, \\
x_{(ij)}x_{(kl)} - x_{(kl)}x_{(ij)} &= 0, && i,j,k,l \in \mathbb{I}_m, \text{ todos distintos}, \\
x_{(jk)}x_{(ik)} - x_{(ij)}x_{(jk)} + x_{(ik)}x_{(ij)} &= 0, && i<j<k \in \mathbb{I}_m, \\
x_{(ik)}x_{(jk)} - x_{(jk)}x_{(ij)} + x_{(ij)}x_{(ik)} &= 0, && i<j<k \in \mathbb{I}_m.
\end{align*}
$$

## El álgebra $\mathcal{FK}_3$


Para $m=3$, el álgebra $\mathcal{FK}_3$ está generada por
$$
x_{(12)},\quad x_{(13)},\quad x_{(23)}
$$
y las relaciones son:
$$
\begin{aligned}
x_{(12)}^2 &= 0, \\
x_{(13)}^2 &= 0, \\
x_{(23)}^2 &= 0, \\
x_{(23)}x_{(13)} - x_{(12)}x_{(23)} + x_{(13)}x_{(12)} &= 0, \\
x_{(13)}x_{(23)} - x_{(23)}x_{(12)} + x_{(12)}x_{(13)} &= 0.
\end{aligned}
$$

## Relaciones en el álgebra $\mathcal{FK}_3$

### *Nilpotencia cuadrática*:
$$
\begin{align*}
x_{(12)}^2 = x_{(13)}^2 = x_{(23)}^2 = 0,
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
x_{12}^2 = x_{13}^2 = x_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
x^{2}_{(ij)} \;=\; x_{(ij)}x_{(ik)} \;+\; x_{(jk)}x_{(ij)} \;+\; x_{(ik)}x_{(jk)}
\;=\;
x_{(ik)}x_{(ij)} \;+\; x_{(ij)}x_{(jk)} \;+\; x_{(jk)}x_{(ik)}
\;=\; 0,
\quad \text{(dada por } B(V)).
$$

$$
\begin{aligned}
x_{(12)}x_{(13)} + x_{(23)}x_{(12)} + x_{(13)}x_{(23)} 
\quad \text{y} \quad 
x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.
\end{aligned}
$$

```gap
###############################################################################
# x_(ij)^2 = 0, x_(ij)x_(jk)+x_(jk)x_(ik)+x_(ik)x_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3      := [ (1,2), (1,3), (2,3) ];;
generators   := gens{[2..4]};;
relationsFK3 := List( generators, s -> s^2);;
i            := 1;;
j            := 2;;
k            := 3;;
Xij          := generators[Position(gensFK3, (i,j))];;
Xik          := generators[Position(gensFK3, (i,k))];;
Xjk          := generators[Position(gensFK3, (j,k))];;
# primera relación: x_(ij)x_(jk) + x_(jk)x_(ik) + x_(ik)x_(ij) = 0
Add(relationsFK3, Xij*Xjk + Xjk*Xik + Xik*Xij );
# segunda relación: x_(jk)x_(ij) + x_(ik)x_(jk) + x_(ij)x_(ik) = 0
Add(relationsFK3, Xjk*Xij + Xik*Xjk + Xij*Xik );
```

```gap
relsFK3 := GP2NPList(relationsFK3);;
PrintNPList( relsFK3 );
#X12^2 
#X13^2 
#X23^2 
#X12X23 + X23X13 + X13X12 
#X23X12 + X13X23 + X12X13 
```

## Relaciones en el álgebra $\mathcal{FK}_3$ dual

### *Nilpotencia cuadrática*:
$$
\begin{align*}
y_{(12)}^2 = y_{(13)}^2 = y_{(23)}^2 = 0.\\
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
y_{12}^2 = y_{13}^2 = y_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
\begin{align*}
    y_{(23)}y_{(13)} + y_{(12)}y_{(23)} + y_{(13)}y_{(12)} &= 0,&&\Rightarrow& y_3y_1 + y_0y_3 + y_1y_0 &= 0, \\
    y_{(13)}x_{(23)} + y_{(23)}y_{(12)} + y_{(12)}y_{(13)} &= 0,&&\Rightarrow& y_1y_3 + y_3y_0 + y_0y_1 &= 0.
\end{align*}
$$

```gap
###############################################################################
# y_(ij)^2 = 0, y_(ij)y_(jk)+y_(jk)y_(ik)+y_(ik)y_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3     := [ (1,2), (1,3), (2,3) ];;
relsFK3dual := [];;
# Paso 1: Relaciones cuadráticas x_(ij)^2 = 0
for k in [1..Length(gensFK3 )] do
    Add(relsFK3dual, [[ [yIndex[k], yIndex[k]] ], [1]]);;
od;
# Paso 2: Relaciones cúbicas de tres índices distintos
triples := [[1,2,3]];;  # único triple distinto para n=3
for t in triples do
    i := t[1];
    j := t[2];
    k := t[3];
    ij := Position(gensFK3, (i,j));
    ik := Position(gensFK3, (i,k));
    jk := Position(gensFK3, (j,k));
    # primera relación: y_(ij)y_(jk) + y_(jk)y_(ik) + y_(ik)y_(ij) = 0
    Add(relsFK3dual,[[[yIndex[ij],yIndex[jk]],[yIndex[jk],yIndex[ik]],[yIndex[ik],yIndex[ij]]],[1,1,1]]);
    # segunda relación: y_(jk)y_(ij) + y_(ik)y_(jk) + y_(ij)y_(ik) = 0
    Add(relsFK3dual,[[[yIndex[jk],yIndex[ij]],[yIndex[ik],yIndex[jk]],[yIndex[ij],yIndex[ik]]],[1,1,1]]);
od;
```

## Calculo de $y_{top}x_{top}$

Las relaciones definitorias del álgebra de Nichols $\mathfrak{B}(V)$ son
$$
x_{(12)}^{2}, \quad x_{(13)}^{2}, \quad x_{(23)}^{2},
$$
$$
x_{(12)}x_{(13)} + x_{(23)}x_{(12)} + x_{(13)}x_{(23)} 
\quad \text{y} \quad 
x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.
$$

Denotamos por $\mathbb{B}$ a la base de $\mathfrak{B}(V)$, la cual se obtiene eligiendo un elemento por cada fila de la siguiente lista y multiplicándolos de arriba hacia abajo, véase *e.g.* [7]:

$$
\begin{array}{l}
1, \, x_{(12)}, \\[4pt]
1, \, x_{(13)}, \, x_{(13)}x_{(12)}, \\[4pt]
1, \, x_{(23)}.
\end{array}
$$

Sea $\mathbb{B}^n = \mathbb{B} \cap \mathfrak{B}^n(V)$, $n \geq 0$. 
El elemento de grado máximo en $\mathbb{B}$ es
$$
x_{\text{top}} = x_{(12)}x_{(13)}x_{(12)}x_{(23)} \in \mathbb{B}^4.
$$

```gap
ActionMatrixByGenerator := function( BaseNichols, J_Nichols, gen )
    local N, Amat, Lt, genGi, bj, pol, mon, PolG2, PolG2Norm, pos, coeffss, bi;
    N := Length(BaseNichols);
    Amat := NullMat(N, N);
    # Precálculos fuera del bucle
    Lt := List(QGNAG.Config.gensA, ElementFq);
    genGi := QGNAG.Config.elmsG[Position( allPairsInG, gen )];
    for bj in [1..N] do
        pol := BaseNichols[bj];
        mon := pol[1];
        PolG2     := [[List(mon[1], m -> Position(Lt, genGi * QGNAG.Config.gensA[m]))], [1]];
        PolG2Norm := StrongNormalFormNP(PolG2, J_Nichols);
        pos       := List(PolG2Norm[1], tx -> Position(BaseNichols, [[tx], [1]]));
        coeffss   := PolG2Norm[2] * ChiForSemidirectProduct( genGi )^Length(pol[1][1]);
        for bi in pos do
            Amat[bi][bj] := coeffss[Position(pos, bi)];
        od;
    od;
    return Amat;
end;
```

module $M_k$.

```gap
QGNAG_QuantumPolynomialsToRecord:=function( data_record_data_decomp, epsilon_and_coeffs )
    local epsilon,
          coefficients,
          data,
          polynomial,
          theta,
          numShifts,
          shiftData,
          shiftCounts,
          index_s,
          p,
          d,
          k,
          coeff;

    theta      := Length(epsilon_and_coeffs);    # Number of simples
    numShifts  := Length(epsilon_and_coeffs[1].coefficients);
    shiftData  := []; 
    for p in [1..theta] do     # Number of admissible shifts for each p.
        shiftData[p] := QGNAG_AllShiftedVectorsLeft( [ data_record_data_decomp[p] ] );
    od;
    shiftCounts := List( shiftData, Length );
    Print("Length(nu) = ", numShifts, "\n");
    Print("Shift counts = ", shiftCounts, "\n");
    # ============================================================
    # Initialize every polynomial directly as [0].
    #
    # data[j].p represents t_{j,p}(q).
    # ============================================================
    data := [];
    
    for index_s in [1..theta] do
        data[index_s] := rec();
        for p in [1..theta] do       # Every t_{j,p} starts as the zero polynomial.
            data[index_s].(String(p)) := [0];
        od;
    od;

    for index_s in [1..theta] do # Compute the coefficients f_{p,d}^j.
        epsilon      := epsilon_and_coeffs[index_s].vector;
        coefficients := epsilon_and_coeffs[index_s].coefficients;
        if Length(coefficients) <> numShifts then
             Error( "Unexpected number of coefficients: ", Length(coefficients), " instead of ", numShifts );
        fi;
        k := 0; # Map the flattened coefficient vector to (p,d).
        for p in [1..theta] do
            for d in [0..shiftCounts[p]-1] do
                k := k + 1;
                coeff := coefficients[k];
                if coeff <> 0 then
                    polynomial := data[index_s].(String(p));
                    # ------------------------------------------------
                    # Extend the coefficient vector only when
                    # the required degree is larger than its
                    # current length.
                    #
                    # For example:
                    #
                    #     [0] + coefficient at d=1
                    #          -> [0,coeff]
                    #
                    #     [0,coeff] + coefficient at d=3
                    #          -> [0,coeff,0,coeff]
                    # ------------------------------------------------
                    while Length(polynomial) < d + 1 do
                        Add(polynomial, 0);
                    od;
                    polynomial[d + 1] := polynomial[d + 1] + coeff;
                    data[index_s].(String(p)) := polynomial;
                fi;
            od;
        od;
    od;
    return data;
end;
```

```gap
# Paso 1: negación pura, record por record, sin ningún baseKey
QGNAG_NegateRecordKeysTest:= function( rec_info )
    local recordNames, recordName, newKey, shiftedRecord;
    recordNames := RecNames( rec_info );
    shiftedRecord := rec();
    for recordName in recordNames do
        newKey := -Int(recordName);
        shiftedRecord.(String(newKey)) := rec_info.(recordName);
    od;
    return shiftedRecord;
end;

# Paso 2: dado un GRUPO ya negado (lista de records), trasladar todos
# por la misma constante para que la clave de soporte más cercana a 0
# (el máximo del soporte, entre valores negativos) pase a valer 0.
QGNAG_ShiftRecordKeysToZeroTest:=function( group )
    local allDegrees, supportDegrees, maxSupportDegree, shiftAmount,
          r, recordNames, recordName, newKey, shiftedRecord, shiftedGroup;

    allDegrees := List( RecNames( group[1] ), Int );
    supportDegrees := Filtered( allDegrees,
        d -> Sum( List( group, x -> x.(String(d)) ) ) > 0 );

    maxSupportDegree := Maximum( supportDegrees );
    shiftAmount := -maxSupportDegree;

    shiftedGroup := [];
    for r in group do
        recordNames := RecNames( r );
        shiftedRecord := rec();
        for recordName in recordNames do
            newKey := Int(recordName) + shiftAmount;
            shiftedRecord.(String(newKey)) := r.(recordName);
        od;
        Add( shiftedGroup, shiftedRecord );
    od;

    return shiftedGroup;
end;

# Wrapper: combina ambos pasos para sign=-1; para sign=1 no hace falta tocar nada
QGNAG_ShiftDecompositionKeysTest:=function( data_record_data_decomp, sign )
    local shiftedData, s, negatedGroup;

    shiftedData := [];
    for s in data_record_data_decomp do
        if sign = -1 then
            negatedGroup := List( s, r -> QGNAG_NegateRecordKeysTest( r ) );
            Add( shiftedData, QGNAG_ShiftRecordKeysToZeroTest( negatedGroup ) );
        else
            Add( shiftedData, s );
        fi;
    od;

    return shiftedData;
end;
```
