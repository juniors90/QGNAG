# El álgebra $\mathfrak{D}=\mathcal{D}(\mathcal{FK}_m\# \Bbbk\mathbb{S}_{3})$
Sea $\mathcal{FK}_m$ el álgebra presentada por generadores

$$
\{ x_{(ij)} \mid i<j \in \mathbb{I}_m \}
$$
y relaciones
$$
\begin{align*}
x_{(ij)}^2 &= 0, && i<j \in \mathbb{I}_m, \\
x_{(ij)}x_{(kl)} - x_{(kl)}x_{(ij)} &= 0, && i,j,k,l \in \mathbb{I}_m, \text{ todos distintos}, \\
x_{(jk)}x_{(ik)} - x_{(ij)}x_{(jk)} + x_{(ik)}x_{(ij)} &= 0, && i<j<k \in \mathbb{I}_m, \\
x_{(ik)}x_{(jk)} - x_{(jk)}x_{(ij)} + x_{(ij)}x_{(ik)} &= 0, && i<j<k \in \mathbb{I}_m.
\end{align*}
$$

## El álgebra $\mathcal{FK}_3$


Para $m=3$, el álgebra $\mathcal{FK}_3$ está generada por
$$
x_{(12)},\quad x_{(13)},\quad x_{(23)}
$$
y las relaciones son:
$$
\begin{aligned}
x_{(12)}^2 &= 0, \\
x_{(13)}^2 &= 0, \\
x_{(23)}^2 &= 0, \\
x_{(23)}x_{(13)} - x_{(12)}x_{(23)} + x_{(13)}x_{(12)} &= 0, \\
x_{(13)}x_{(23)} - x_{(23)}x_{(12)} + x_{(12)}x_{(13)} &= 0.
\end{aligned}
$$

## Relaciones en el álgebra $\mathcal{FK}_3$

### *Nilpotencia cuadrática*:
$$
\begin{align*}
x_{(12)}^2 = x_{(13)}^2 = x_{(23)}^2 = 0,
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
x_{12}^2 = x_{13}^2 = x_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
x^{2}_{(ij)} \;=\; x_{(ij)}x_{(ik)} \;+\; x_{(jk)}x_{(ij)} \;+\; x_{(ik)}x_{(jk)}
\;=\;
x_{(ik)}x_{(ij)} \;+\; x_{(ij)}x_{(jk)} \;+\; x_{(jk)}x_{(ik)}
\;=\; 0,
\quad \text{(dada por } B(V)).
$$

$$
\begin{aligned}
x_{(12)}x_{(13)} + x_{(23)}x_{(12)} + x_{(13)}x_{(23)} 
\quad \text{y} \quad 
x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.
\end{aligned}
$$

```gap
###############################################################################
# x_(ij)^2 = 0, x_(ij)x_(jk)+x_(jk)x_(ik)+x_(ik)x_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3      := [ (1,2), (1,3), (2,3) ];;
generators   := gens{[2..4]};;
relationsFK3 := List( generators, s -> s^2);;
i            := 1;;
j            := 2;;
k            := 3;;
Xij          := generators[Position(gensFK3, (i,j))];;
Xik          := generators[Position(gensFK3, (i,k))];;
Xjk          := generators[Position(gensFK3, (j,k))];;
# primera relación: x_(ij)x_(jk) + x_(jk)x_(ik) + x_(ik)x_(ij) = 0
Add(relationsFK3, Xij*Xjk + Xjk*Xik + Xik*Xij );
# segunda relación: x_(jk)x_(ij) + x_(ik)x_(jk) + x_(ij)x_(ik) = 0
Add(relationsFK3, Xjk*Xij + Xik*Xjk + Xij*Xik );
relsFK3 := GP2NPList(relationsFK3);;
```

## Relaciones en el álgebra $\mathcal{FK}_3$ dual

### *Nilpotencia cuadrática*:
$$
\begin{align*}
y_{(12)}^2 = y_{(13)}^2 = y_{(23)}^2 = 0.\\
\end{align*}
$$
Para nosotros serán:
$$
\begin{align*}
y_{12}^2 = y_{13}^2 = y_{23}^2 = 0.
\end{align*}
$$

### Relaciones triples para $i<j<k$:

- Para $1<2<3$:
$$
\begin{align*}
    y_{(23)}y_{(13)} + y_{(12)}y_{(23)} + y_{(13)}y_{(12)} &= 0,&&\Rightarrow& y_3y_1 + y_0y_3 + y_1y_0 &= 0, \\
    y_{(13)}x_{(23)} + y_{(23)}y_{(12)} + y_{(12)}y_{(13)} &= 0,&&\Rightarrow& y_1y_3 + y_3y_0 + y_0y_1 &= 0.
\end{align*}
$$

```gap
###############################################################################
# y_(ij)^2 = 0, y_(ij)y_(jk)+y_(jk)y_(ik)+y_(ik)y_(ij)=0, etc. 
# (álgebra Fomin-Kirillov n=3,
###############################################################################

# Generadores x_(ij)
gensFK3     := [ (1,2), (1,3), (2,3) ];;
relsFK3dual := [];;
# Paso 1: Relaciones cuadráticas x_(ij)^2 = 0
for k in [1..Length(gensFK3 )] do
    Add(relsFK3dual, [[ [yIndex[k], yIndex[k]] ], [1]]);;
od;
# Paso 2: Relaciones cúbicas de tres índices distintos
triples := [[1,2,3]];;  # único triple distinto para n=3
for t in triples do
    i := t[1];
    j := t[2];
    k := t[3];
    ij := Position(gensFK3, (i,j));
    ik := Position(gensFK3, (i,k));
    jk := Position(gensFK3, (j,k));
    # primera relación: y_(ij)y_(jk) + y_(jk)y_(ik) + y_(ik)y_(ij) = 0
    Add(relsFK3dual,[[[yIndex[ij],yIndex[jk]],[yIndex[jk],yIndex[ik]],[yIndex[ik],yIndex[ij]]],[1,1,1]]);
    # segunda relación: y_(jk)y_(ij) + y_(ik)y_(jk) + y_(ij)y_(ik) = 0
    Add(relsFK3dual,[[[yIndex[jk],yIndex[ij]],[yIndex[ik],yIndex[jk]],[yIndex[ij],yIndex[ik]]],[1,1,1]]);
od;
```

## Calculo de $y_{top}x_{top}$

Las relaciones definitorias del álgebra de Nichols $\mathfrak{B}(V)$ son
$$

$$
$$

x_{(13)}x_{(12)} + x_{(12)}x_{(23)} + x_{(23)}x_{(13)}.

Denotamos por $\mathbb{B}$ a la base de $\mathfrak{B}(V)$, la cual se obtiene eligiendo un elemento por cada fila de la siguiente lista y multiplicándolos de arriba hacia abajo, véase *e.g.* [7]:

\begin{array}{l}
1, \, x_{(12)}, \\[4pt]
1, \, x_{(13)}, \, x_{(13)}x_{(12)}, \\[4pt]
1, \, x_{(23)}.
\end{array}
$$

Sea $\mathbb{B}^n = \mathbb{B} \cap \mathfrak{B}^n(V)$, $n \geq 0$. 
El elemento de grado máximo en $\mathbb{B}$ es
$$
x_{\text{top}} = x_{(12)}x_{(13)}x_{(12)}x_{(23)} \in \mathbb{B}^4.

## Vectors $\nu_{i,j}$

Let $\operatorname{char}(\mathcal{L}_i)=\sum_{d=0}^{\theta}\sum_{j=1}^{N}m_{j,d}\,M_j[d],$
where $m_{j,d}\in\mathbb{N}$ denotes the multiplicity of the simple module
$M_j$ in homogeneous degree $d$.

For each degree $d$, define the block vector
$L_d=(m_{1,d},m_{2,d},\ldots,m_{N,d})\in\mathbb{Q}^{N}.$

The function $\texttt{QGNAG\_CharToVector}$ associates to the graded character of
$\mathcal{L}_i$ the vector
$\nu_{i,0}=(L_0,L_1,\ldots,L_{\theta})\in\mathbb{Q}^{N(\theta+1)},$
obtained by concatenating the multiplicity vectors of each homogeneous degree.

To encode all cyclic shifts of the grading, we define, for every
$k\in\{0,\ldots,\theta\}$,
$\nu_{i,k}=(L_{\theta-k+1},\ldots,L_{\theta},L_0,\ldots,L_{\theta-k}),$
where the indices are understood modulo $\theta+1$.

Equivalently, $\nu_{i,k}=\bigl(L_{(-k)\bmod(\theta+1)},L_{(1-k)\bmod(\theta+1)},\ldots,L_{(\theta-k)\bmod(\theta+1)}\bigr).$

Hence,
\begin{align*}
\nu_{i,0}
    &= (L_0,L_1,\ldots,L_{\theta}),\\
\nu_{i,1}
    &= (L_{\theta},L_0,L_1,\ldots,L_{\theta-1}),\\
\nu_{i,2}
    &= (L_{\theta-1},L_{\theta},L_0,\ldots,L_{\theta-2}),\\
&\ \vdots\\
\nu_{i,\theta}
    &= (L_1,L_2,\ldots,L_{\theta},L_0).
\end{align*}

The function $\texttt{QGNAG\_CharToVector}$ computes the vector
$\nu_{i,0}$, while the function
$\texttt{QGNAG\_CharToShiftedVectors}$ returns the family
$
\{\nu_{i,0},\nu_{i,1},\ldots,\nu_{i,\theta}\},
$
consisting of all cyclic permutations of the block decomposition of the graded
character.

```gap
QGNAG_CharToVector := function( record_data_decomp )
    local N,
          degrees,
          ell,
          vector,
          d,
          i,
          degree;
    
    N       := Length(record_data_decomp);
    degrees := List(RecNames(record_data_decomp[1]), x -> Int(x));
    Sort(degrees);
    ell := Maximum(degrees);
    vector := [];
    for d in [0..ell] do
        degree := String(d);
        for i in [1..N] do
            Add(vector, record_data_decomp[i].(degree));
        od;
    od;
    return vector;
end;
```

```gap
QGNAG_CharToGradedVectors := function(record_data_decomp)
    local N,       #
          degrees, #
          degree,  #
          ell,     #
          i,       #
          vectors, #
          vector;  #
    
    N := Length(record_data_decomp);
    degrees := List(
        RecNames(record_data_decomp[1]),
        x -> Int(x)
    );
    Sort(degrees);
    ell := Maximum(degrees);
    vectors := [];
    for degree in [0..ell] do
        vector := [];
        for i in [1..N] do
            Add(vector, record_data_decomp[i].(degree));
        od;
        Add(vectors, vector);
    od;
    return vectors;
end;
```

```gap
QGNAG_CharToShiftedVector := function( record_data_decomp, ell )
    local N,
          degrees,
          theta,
          vector,
          d,
          shifted_degree,
          i,
          degree;
    
    N       := Length(record_data_decomp);
    degrees := List(RecNames(record_data_decomp[1]), x -> Int(x));
    theta   := Maximum(degrees);
    ell     := ell mod (theta + 1);
    vector  := [];
    for d in [0..theta] do
        shifted_degree := (d - ell) mod (theta + 1);
        degree         := String(shifted_degree);
        for i in [1..N] do
            Add(vector, record_data_decomp[i].(degree));
        od;
    od;
    return vector;
end;
```

```gap
QGNAG_CharToShiftedGradeedVector := function( record_data_decomp, j )
    local N,
          degrees,
          degree,
          ell,
          shifted_vector,
          d,
          source_degree,
          block,
          i;

    N              := Length(record_data_decomp);
    degrees        := List(RecNames(record_data_decomp[1]), x -> Int(x));
    ell            := Maximum(degrees);
    shifted_vector := [];
    for d in [0..ell] do
        source_degree := (d - j) mod (ell + 1);
        degree        := String(source_degree);
        block := [];
        for i in [1..N] do
            Add(block, record_data_decomp[i].(degree));
        od;
        Add(shifted_vector, block);
    od;
    return shifted_vector;
end;
```

Let
$
\mathcal{E}=
(e_{1,1}^{0},\ldots,e_{1,N}^{0},
e_{1,1}^{1},\ldots,e_{1,N}^{1},
\ldots,
e_{1,1}^{\ell},\ldots,e_{1,N}^{\ell})
$
be the standard graded basis, where the coordinates are ordered first by
degree and then by simple module. For each $i=1,\ldots,\theta$ and
$j=0,\ldots,\ell$, let $\nu_{i,j}$ be the shifted vector defined above,
and consider the ordered family
$
\mathcal{B}=
(\nu_{1,0},\ldots,\nu_{1,\ell},
 \nu_{2,0},\ldots,\nu_{2,\ell},
 \ldots,
 \nu_{\theta,0},\ldots,\nu_{\theta,\ell}).
$

Assuming that $\mathcal{B}$ is a basis of $V=\mathbb{Q}^{N(\ell+1)}$, construct the matrix
$
P=\begin{pmatrix}|&|&&|\\
\nu_{1,0}&\nu_{1,1}&\cdots&\nu_{\theta,\ell}\\
|&|&&|
\end{pmatrix}_{\mathcal{E}},
$
whose columns are the vectors of $\mathcal{B}$ written in the standard
basis $\mathcal{E}$. Thus, for
$\varepsilon_i=
(d_{i1}^{0},\ldots,d_{iN}^{0},
 d_{i1}^{1},\ldots,d_{iN}^{1},
 \ldots,
 d_{i1}^{\ell},\ldots,d_{iN}^{\ell})^{T},
$
its coordinates in the basis $\mathcal{B}$ are given by
$[\varepsilon_i]_{\mathcal{B}}=P^{-1}[\varepsilon_i]_{\mathcal{E}}.$
Equivalently,
$$
\varepsilon_i=
\sum_{r=1}^{\theta}\sum_{j=0}^{\ell}
c_{r,j}\nu_{r,j},
\,
[\varepsilon_i]_{\mathcal{B}}
=
(c_{1,0},\ldots,c_{1,\ell},
 c_{2,0},\ldots,c_{\theta,\ell})^{T}.
$$
If all vectors $\varepsilon_i$ are assembled as columns of a matrix
$$
E=
\begin{pmatrix}
|&|&&|\\
\varepsilon_1&\varepsilon_2&\cdots&\varepsilon_{\theta(\ell+1)}\\
|&|&&|
\end{pmatrix}_{\mathcal{E}},
$$
then the change-of-basis matrix is obtained simultaneously as $C=P^{-1}E$,
where the $i$-th column of $C$ contains the coordinates of
$\varepsilon_i$ with respect to $\mathcal{B}$.

```gap
QGNAG_PrintQP := function( coeffs, names )
    local terms, i, c, s, result;
    terms := [];
    for i in [1 .. Length(coeffs)] do
        c := coeffs[i];
        if c <> 0 then
            if c = 1 then
                s := names[i];
            elif c = -1 then
                s := Concatenation("-", names[i]);
            else
                s := Concatenation(String(c), " ", names[i]);
            fi;
            Add(terms, s);
        fi;
    od;
    
    if Length(terms) = 0 then
        result := "\\epsilon = 0";
    else
        result := "\\epsilon = ";
        for i in [1 .. Length(terms)] do
            if i = 1 then
                result := Concatenation(result, terms[i]);
            else
                if terms[i][1] = '-' then
                    result := Concatenation(result, " - ", terms[i]{[2 .. Length(terms[i])]});
                else
                    result := Concatenation(result, " + ", terms[i]);
                fi;
            fi;
        od;
    fi;
    Print(result, "\n");
    return result;
end;;
```

```gap
QGNAG_QPToLaTeX:=function( coeffs, names, index )
    local terms, i, c, s, result;

    terms := [];
    for i in [1..Length(coeffs)] do
        c := coeffs[i];
        if c <> 0 then
            if c = 1 then
                s := names[i];
            elif c = -1 then
                s := Concatenation("-", names[i]);
            else
                s := Concatenation(String(c), "\\,", names[i]);
            fi;
            Add(terms, s);
        fi;
    od;
    if Length(terms) = 0 then
        result := Concatenation("\\epsilon_", String(index), " = 0");
    else
        result := Concatenation("\\epsilon_", String(index), " = ");
        for i in [1..Length(terms)] do
            if i = 1 then
                result := Concatenation(result, terms[i]);
            else
                if terms[i][1] = '-' then
                    result := Concatenation(
                        result,
                        " - ",
                        terms[i]{[2..Length(terms[i])]}
                    );
                else
                    result := Concatenation(
                        result,
                        " + ",
                        terms[i]
                    );
                fi;
            fi;
        od;
    fi;
    return result;
end;
```

```gap
QGNAG_SaveQPToLaTeX:=function(filename, coeffs, names, index)
    local out, result;
    out := OutputTextFile(filename, false);
    result := QGNAG_QPToLaTeX(coeffs, names, index);
    AppendTo(out, "\\[\n");
    AppendTo(out, result);
    AppendTo(out, "\n\\]\n");
    CloseStream(out);
end;
```

```gap
QGNAG_EpsilonPolynomials := function( data_record_data_decomp, SimplesMn, rec_info_nichols)
    local nu,
          A,
          result,
          index,
          epsilon,
          coefficients,
          N,
          j,
          number_of_shifts,
          polynomial,
          d,
          position,
          coeff,
          epsilon_data;

    nu := QGNAG_AllShiftedVectorsLeft( data_record_data_decomp );
    A := TransposedMat(nu);
    N := Length(data_record_data_decomp);
    result := [];
    for index in [1..N] do
        epsilon := QGNAG_GradedFusionVector(
            index,
            SimplesMn,
            rec_info_nichols
        );
        if RankMat(A) <> RankMat( TransposedMat( Concatenation(nu, [epsilon]) ) ) then
            Error( "epsilon_", index, " is not in the span of the shifted vectors" );
        fi;
        coefficients := SolutionMat( TransposedMat(A), epsilon );
        epsilon_data := rec(
            epsilon := index,
            coefficients := coefficients,
            polynomials := []
        );
        position := 1;
        for j in [1..N] do
            number_of_shifts := Length( QGNAG_AllCharToShiftedVectorsLeft( data_record_data_decomp[j] ) );
            polynomial := [];
            for d in [0..number_of_shifts-1] do
                coeff := coefficients[position];
                Add(polynomial, coeff);
                position := position + 1;
            od;
            Add( epsilon_data.polynomials, polynomial );
        od;
        Add(result, epsilon_data);
    od;
    return result;
end;
```

```gap
QGNAG_EvaluateEpsilonPolynomialsAtOne:=function( epsilon_polynomials )
    local result, epsilon_data, values, polynomial;

    result := [];

    for epsilon_data in epsilon_polynomials do
        values := [];

        for polynomial in epsilon_data.polynomials do
            Add(values, Sum(polynomial));
        od;

        Add(
            result,
            rec(
                epsilon := epsilon_data.epsilon,
                values := values
            )
        );
    od;

    return result;
end;
```
