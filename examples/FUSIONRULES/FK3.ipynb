# El álgebra $\mathfrak{D}=\mathcal{D}(\mathcal{FK}_m\# \Bbbk\mathbb{S}_{3})$
Sea $\mathcal{FK}_m$ el álgebra presentada por generadores

$$
\{ x_{(ij)} \mid i<j \in \mathbb{I}_m \}
$$
y relaciones
$$
\begin{align*}
x_{(ij)}^2 &= 0, && i<j \in \mathbb{I}_m, \\
x_{(ij)}x_{(kl)} - x_{(kl)}x_{(ij)} &= 0, && i,j,k,l \in \mathbb{I}_m, \text{ todos distintos}, \\
x_{(jk)}x_{(ik)} - x_{(ij)}x_{(jk)} + x_{(ik)}x_{(ij)} &= 0, && i<j<k \in \mathbb{I}_m, \\
x_{(ik)}x_{(jk)} - x_{(jk)}x_{(ij)} + x_{(ij)}x_{(ik)} &= 0, && i<j<k \in \mathbb{I}_m.
\end{align*}
$$

## El álgebra $\mathcal{FK}_3$


Para $m=3$, el álgebra $\mathcal{FK}_3$ está generada por
$$
x_{(12)},\quad x_{(13)},\quad x_{(23)}
$$
y las relaciones son:
$$
\begin{aligned}
x_{(12)}^2 &= 0, \\
x_{(13)}^2 &= 0, \\
x_{(23)}^2 &= 0, \\
x_{(23)}x_{(13)} - x_{(12)}x_{(23)} + x_{(13)}x_{(12)} &= 0, \\
x_{(13)}x_{(23)} - x_{(23)}x_{(12)} + x_{(12)}x_{(13)} &= 0.
\end{aligned}
$$

We consider the group
$$
\Gamma=C_3\rtimes C_2
=\langle r,s\mid r^3=s^2=1,\ srs=r^{-1}\rangle.
$$

Its elements are
$
\Gamma=\{1,r,r^2,s,rs,r^2s\}.
$

For every $g\in\Gamma$, its conjugacy class is defined by
$$
\mathcal O_g=\{xgx^{-1}\mid x\in\Gamma\}.
$$

Since
$$
\Gamma=\{1,r,r^2,s,rs,r^2s\},
$$
and every element can be written uniquely as $r^is^j$, with
$i\in\{0,1,2\}$ and $j\in\{0,1\}$, the conjugacy classes can be computed using the relation
$$
srs=r^{-1}.
$$

First,
$$
x1x^{-1}=1,\qquad \forall\,x\in\Gamma,
$$
hence
$$
\mathcal O_1=\{1\}.
$$
Now consider $r$. Since powers of $r$ commute,
$$
r^i r r^{-i}=r,\qquad i=0,1,2.
$$
On the other hand,
$$
srs=r^{-1}=r^2,
$$
and therefore
$$
(sr^i)r(sr^i)^{-1}
=s(r^ir^{-i})s
=srs
=r^2.
$$
Thus the only conjugates of $r$ are $r$ and $r^2$, so that
$$
\mathcal O_r=\{r,r^2\}.
$$

Finally, let us compute the conjugates of $s$. Since
$$
r^isr^{-i}=r^{2i}s,
$$
we obtain
$$
rsr^{-1}=r^2s,
$$
and
$$
r^2sr^{-2}=rs.
$$
Moreover,
$$
sss^{-1}=s,
$$
and
$$
(sr^i)s(sr^i)^{-1}
=s(r^isr^{-i})s
=s(r^{2i}s)s
=r^is,
$$
which produces no new elements. Hence
$$
\mathcal O_s=\{s,rs,r^2s\}.
$$

Therefore the conjugacy classes of $\Gamma$ are
$$
\mathcal O_1=\{1\},\qquad
\mathcal O_r=\{r,r^2\},\qquad
\mathcal O_s=\{s,rs,r^2s\}.
$$

Consequently, we may choose
$$
Q=\{1,r,s\}
$$
as a complete set of representatives of the conjugacy classes of $\Gamma$.

Hence we choose
$$
\mathcal{Q}=\{1,r,s\}
$$
as a complete set of representatives of conjugacy classes.

The corresponding isotropy (centralizer) subgroups are
$$
\Gamma_1=\Gamma,
$$

$$
\Gamma_r=\langle r\rangle
=\{1,r,r^2\}\cong C_3,
$$
and
$$
\Gamma_s=\langle s\rangle
=\{1,s\}\cong C_2.
$$
The irreducible representations of these groups are as follows.

Since
$$
\widehat{\Gamma}
=\{\mathbf 1,\varepsilon,\pi\},
$$
there are two one-dimensional representations and one irreducible representation of dimension
$2$.

The subgroup
$$
\Gamma_r\cong C_3
$$
has three one-dimensional irreducible representations,
$$
\widehat{\Gamma_r}
=\{\chi_0,\chi_1,\chi_2\},
$$
where, if $\omega$ is a primitive cube root of unity,
$$
\chi_j(r)=\omega^j,
\qquad
j=0,1,2.
$$

Finally,
$$
\Gamma_s\cong C_2
$$
has two one-dimensional irreducible representations,
$$
\widehat{\Gamma_s}
=\{\mathbf 1,\operatorname{sgn}\},
$$
given by
$$
\operatorname{sgn}(s)=-1.
$$
Therefore the simple Yetter-Drinfeld modules over $\Gamma$ are
$$
M(1,\mathbf1),\qquad
M(1,\varepsilon),\qquad
M(1,\pi),
$$
$$
M(r,\chi_0),\qquad
M(r,\chi_1),\qquad
M(r,\chi_2),
$$

$$
M(s,\mathbf1),\qquad
M(s,\operatorname{sgn}).
$$

Their dimensions are

$$
\dim M(1,\rho)=\dim\rho,
$$

$$
\dim M(r,\chi_j)
=[\Gamma:\Gamma_r]\dim\chi_j
=2,
$$
and
$$
\dim M(s,\psi)
=[\Gamma:\Gamma_s]\dim\psi
=3.
$$
Hence the dimensions of the simple Yetter-Drinfeld modules are
$$
1,\;1,\;2,\;2,\;2,\;2,\;3,\;3.
$$

The group $\Gamma$ has a unique irreducible representation of dimension two, which we denote by
$$
\pi:\Gamma\longrightarrow GL(V).
$$
Since $r$ has order $3$, its minimal polynomial divides
$$
x^3-1=(x-1)(x-\omega)(x-\omega^2).
$$
As $\pi$ is irreducible and two-dimensional, the eigenvalue $1$ cannot occur. Hence $r$ has the distinct eigenvalues $\omega$ and $\omega^2$, where
$$
\omega=e^{2\pi i/3}.
$$
Choosing eigenvectors $v_1,v_2$ satisfying
$$
r\cdot v_1=\omega^2v_1,\qquad
r\cdot v_2=\omega v_2,
$$
the relation
$$
srs=r^{-1}
$$
implies
$$
r(sv_i)=\omega^{-i}(sv_i),
$$
that is, $s$ exchanges the two eigenspaces of $r$. Rescaling the basis if necessary, we may assume
$$
s\cdot v_1=v_2,\qquad
s\cdot v_2=v_1.
$$

Therefore,
$$
\pi(r)=
\begin{pmatrix}
\omega^2&0\\
0&\omega
\end{pmatrix},
\qquad
\pi(s)=
\begin{pmatrix}
0&1\\
1&0
\end{pmatrix},
$$
which is precisely the realization returned by the computational package.

## $M(s, \pm)$

The matrices obtained above depend on the choice of generators of the cyclic subgroup
$$
\langle r\rangle\simeq C_3
$$
and on the choice of representatives of the left cosets of $\Gamma_s$.

In our computations we take the basis
$$
\mathcal B=\{\,1\otimes v,\ r\otimes v,\ r^2\otimes v\,\},
$$
so that
$$
r\cdot(1\otimes v)=r\otimes v,\qquad
r\cdot(r\otimes v)=r^2\otimes v,\qquad
r\cdot(r^2\otimes v)=1\otimes v,
$$
which yields
$$
[r]_{\mathcal B}=
\begin{pmatrix}
0&0&1\\
1&0&0\\
0&1&0
\end{pmatrix}.
$$
The GAP implementation chooses the opposite generator of the cyclic subgroup
$\langle r\rangle$, namely $r^{-1}$ instead of $r$. Consequently,
$$
r^{-1}\cdot(1\otimes v)=r^2\otimes v,\qquad
r^{-1}\cdot(r\otimes v)=1\otimes v,\qquad
r^{-1}\cdot(r^2\otimes v)=r\otimes v,
$$
and therefore the matrix returned by GAP is
$$
[r^{-1}]_{\mathcal B}=
\begin{pmatrix}
0&1&0\\
0&0&1\\
1&0&0
\end{pmatrix},
$$
which is the inverse of the matrix representing $r$ above.

Thus, the difference between the matrices is entirely due to the choice of generator of
$\langle r\rangle$; both realizations define isomorphic representations.

# The simple Verma modules

Let us take the expression
$$
\Omega := -2\tau^2\delta_e + 2\tau\sigma\delta_e + 4\sigma\tau\delta_e - 2\tau\delta_e + 2\sigma\delta_e - 2\sigma\tau - 2
4$$. 
Since $\delta_e$ acts as the identity in all cases, we can formally replace
$$
\delta_e \mapsto I.
$$ 
Then
$$
\Omega = -2\tau^2 + 2\tau\sigma + 4\sigma\tau - 2\tau + 2\sigma - 2\sigma\tau - 2.
$$
Grouping:
$$
\Omega = -2\tau^2 + 2\tau\sigma + 2\sigma\tau - 2\tau + 2\sigma - 2.
$$
We will use:
$$
\sigma = (12), \quad \tau = (123).
$$
In $\mathbb{S}_3$,
$$
\tau^2 = (132),
$$
and furthermore
\begin{align*}
\tau\sigma &= (123)(12) = (13), \\
\sigma\tau &= (12) (123) = (23).
\end{align*}
Therefore,
$$
\Omega = -2(132) + 2(13) + 2(23) - 2(123) + 2(12) - 2e.
$$

## The Verma modules $\mathrm{M}(e, +)$ isn't $\mathfrak{D}$-simple.

- 1) trivial representation on $|e\rangle_{+}$

The trivial representation satisfies
$$
\varepsilon(\text{transposition}) = 1, \quad \varepsilon(\text{3-cycle}) = 1.
$$

Then:
\begin{align*}
\varepsilon(\tau) = 1, \quad \varepsilon(\tau^2) = 1, \\
\varepsilon(\sigma) = 1, \quad \varepsilon(\tau\sigma) = 1, \quad \varepsilon(\sigma\tau) = 1.
\end{align*}
We apply this term by term:
\begin{align*}
-2\tau^2 &\mapsto -2(1) = -2, \\
2\tau\sigma &\mapsto 2(1) = 2, \\
2\sigma\tau &\mapsto 2(1) = 2, \\
-2\tau &\mapsto -2(1) = -2, \\
2\sigma &\mapsto 2(1) = 2, \\
-2e &\mapsto -2.
\end{align*}
Adding up:
$$
-2 + 2 + 2 - 2 + 2 - 2 = 0.
$$
Therefore,
$$
\Omega|e\rangle_{+} = 0 |e\rangle_{-} = 0.
$$

The relation is a direct consequence of Schur's Lemma.

Given the decomposition $M \cong \bigoplus_{j=1}^n S_j^{\oplus t_j}$, applying the left-exact functor $\operatorname{Hom}_A(S_i, -)$ yields an isomorphism of vector spaces over the base field $k$:

$$\operatorname{Hom}_A(S_i, M) \cong \bigoplus_{j=1}^n \operatorname{Hom}_A(S_i, S_j)^{\oplus t_j}$$

Because the modules $\{S_1, \dots, S_n\}$ are irreducible and mutually non-isomorphic, Schur's Lemma dictates that $\operatorname{Hom}_A(S_i, S_j) = 0$ whenever $i \neq j$. Consequently, all cross-terms vanish, leaving:

$$\operatorname{Hom}_A(S_i, M) \cong \operatorname{Hom}_A(S_i, S_i)^{\oplus t_i}$$

By Schur's Lemma, $\operatorname{Hom}_A(S_i, S_i)$ is a division algebra over $k$. Taking dimensions over $k$ gives the explicit formula for the multiplicity $t_i$:

$$t_i = \frac{\dim_k \operatorname{Hom}_A(S_i, M)}{\dim_k \operatorname{Hom}_A(S_i, S_i)}$$

If the base field $k$ is algebraically closed, $\operatorname{Hom}_A(S_i, S_i) \cong k$, making its dimension $1$. In this case, the relation reduces entirely to:

$$t_i = \dim_k \operatorname{Hom}_A(S_i, M)$$
