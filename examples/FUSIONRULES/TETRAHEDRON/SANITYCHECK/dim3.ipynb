```gap
K           := DirectProduct(CyclicGroup(2), CyclicGroup(2));;
H           := CyclicGroup(6);;
AutK        := AutomorphismGroup(K);
T           := GroupHomomorphismByImages(H, AutK, [H.1], [List(AutK)[2]]);;
G           := SemidirectProduct(H, T, K);;
embK        := Embedding(G, 2);;
embH        := Embedding(G, 1);;
HinG        := List([0 .. Size(H) - 1], x -> embH(H.1^x));;
allPairsInG := Concatenation(List(K, x -> List(HinG, y -> embK(x) * y)));;
```

```gap
RepresentationM := function(G, a, b, c)
    local A1, A2, A3, A4, gens, mgens, rho;

    if not (a in [0,1] and b in [0,1] and c in [0,1]) then
        Error("The parameters a, b and c must belong to {0,1}.");
    fi;

    A1 := [ [ 0, 0, (-1)^c ],
            [ 1, 0, 0 ],
            [ 0, 1, 0 ] ];

    A2 := A1^2;

    A3 := [ [ (-1)^(a+b), 0, 0 ],
            [ 0, (-1)^a,   0 ],
            [ 0, 0, (-1)^b ] ];

    A4 := [ [ (-1)^a,     0, 0 ],
            [ 0, (-1)^b,   0 ],
            [ 0, 0, (-1)^(a+b) ] ];

    gens  := GeneratorsOfGroup(G);
    mgens := [A1, A2, A3, A4];

    rho := GroupHomomorphismByImages(
        G,
        Group(mgens),
        gens,
        mgens
    );
    return rho;
end;
```

```gap
CheckRelations := function(simple)
    local Id, A1, A2, A3, A4, G;
    if not (a in [0,1] and b in [0,1] and c in [0,1]) then
        Error("The parameters a, b and c must belong to {0,1}.");
    fi;
    G  := Source(simple);
    A1 := simple(G.1);
    A2 := simple(G.2);
    A3 := simple(G.3);
    A4 := simple(G.4);
    Id := IdentityMat(Length(A4));
    return [
        A3^2             = Id,
        A4^2             = Id,
        A2^3             = Id,
        A1^2*A2^2        = Id,
        A1*A4*A1^5*A3    = Id,
        (A3*A4)^2        = Id,
        A3*A1*A3*A1^5*A4 = Id
    ];
end;
```
