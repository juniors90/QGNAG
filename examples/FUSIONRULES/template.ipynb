## Yetter-Drinfeld category over finite groups.
Let $G$ be a finite group.
A Yetter-Drinfeld module over the group algebra $\Bbbk G$ is a $G$-graded vector space
$$
V=\bigoplus_{g\in G}V_g
$$
which is also a left $\Bbbk G$-module such that
$$
h\cdot V_g \subseteq V_{hgh^{-1}}
$$
for all $g,h\in G$. Denote the category of Yetter-Drinfeld modules over $\Bbbk G$ by
$$
{}_G^G\mathcal{YD}.
$$
The category ${}_G^G\mathcal{YD}$ is braided. For $V,W\in {}_G^G\mathcal{YD}$, the braiding
$$
c_{V,W}:V\otimes W\to W\otimes V
$$
is given by
$$
c_{V,W}(v\otimes w)=g\cdot w\otimes v,
\qquad v\in V_g,\; w\in W.
$$

## Yetter-Drinfeld category over finite groups.
Let $G$ be a finite group. A Yetter-Drinfeld module over the group algebra $\Bbbk G$ is a left $\Bbbk G$-module $V$ together with a left $\Bbbk G$-coaction
$$
\rho:V\to \Bbbk G\otimes V
$$
such that
$$
\rho(h\cdot v)=h v_{(-1)} h^{-1}\otimes h\cdot v_{(0)}
$$
for all $h\in G$ and $v\in V$, where
$$
\rho(v)=v_{(-1)}\otimes v_{(0)}.
$$
Equivalently,
$$
V=\bigoplus_{g\in G}V_g,
\qquad
V_g=\{v\in V\mid \rho(v)=g\otimes v\},
$$
and the compatibility condition becomes
$$
h\cdot V_g\subseteq V_{hgh^{-1}}.
$$

Denote this category by ${}_G^G\mathcal{YD}$.

```gap
UpdateSimpleBase := function(simple)
    local m, new_base, elems;

    m := Length(simple.base);
    elems := Elements(GF(4));

    if m = 3 then
        new_base := List([1..m], i ->
            TensorElement(ElementSDP(elems[1], i-1), [1])
        );

        simple.base := new_base;
        return simple;

    elif m = 4 then
        new_base := List([1..m], i ->
            TensorElement(ElementSDP(elems[i], 0), [1])
        );

        simple.base := new_base;
        return simple;

    else
        return simple;
    fi;
end;
```

```gap
RepresentationMdim3 := function(G, a, b, c)
    local A1, A2, A3, A4, gens, mgens, rho;

    if not (a in [0,1] and b in [0,1] and c in [0,1]) then
        Error("The parameters a, b and c must belong to {0,1}.");
    fi;

    A1 := [ [ 0, 0, (-1)^c ],
            [ 1, 0, 0 ],
            [ 0, 1, 0 ] ];

    A2 := A1^2;

    A3 := [ [ (-1)^(a+b), 0, 0 ],
            [ 0, (-1)^a,   0 ],
            [ 0, 0, (-1)^b ] ];

    A4 := [ [ (-1)^a,     0, 0 ],
            [ 0, (-1)^b,   0 ],
            [ 0, 0, (-1)^(a+b) ] ];

    gens  := GeneratorsOfGroup(G);
    mgens := [A1, A2, A3, A4];

    rho := GroupHomomorphismByImages(
        G,
        Group(mgens),
        gens,
        mgens
    );
    return rho;
end;
```

```gap
QGNAG_FusionRulesToLaTeX := function(filename, simplesDim1, SimplesMn, SimpleNames)
    local out, i, j, rule, line, count, n;

    out := OutputTextFile(filename, false);

    n := Length(simplesDim1);

    for i in [1..n] do

        AppendTo(out,
            Concatenation(
                "\\subsubsection{$M_", String(Position(SimplesMn, simplesDim1[i])),
                "\\otimes -$}\n",
                "\\begin{align*}\n"
            )
        );

        line := "";
        count := 0;

        for j in [i..n] do

            rule := QGNAG_FusionRuleToIndexTex(
                        QGNAG_TensorProductOfSimples(
                            simplesDim1[i],
                            simplesDim1[j]
                        ),
                        SimplesMn,
                        SimpleNames
                    );

            # Cambiar \otimes por \ot
            # rule := ReplacedString(rule, "\\otimes", "\\ot");

            #line := Concatenation( line, rule );
            #count := count + 1;
            #if j < n then
            #    line := Concatenation(line, " &&& ");
            #fi;

            line := Concatenation( line, rule );
            count := count + 1;
            
            if count = 4 or j = n then
                AppendTo(out, line, "\\\\\n");
                line := "";
                count := 0;
            else
                line := Concatenation(line, " &&& ");
            fi;

            if count = 4 and j < n then
                line := Concatenation(line, "\\\\\n");
                AppendTo(out, line);
                line := "";
                count := 0;
            fi;

        od;

        if line <> "" then
            line := Concatenation(line, "\n");
            AppendTo(out, line);
        fi;

        AppendTo(out,
            "\\end{align*}\n\n"
        );

    od;

    CloseStream(out);

end;
```
