## Espacio de homomorfismos $\Gamma$-equivariantes

Sean $(\rho, V)$ y $(\tau, W)$ dos representaciones de un grupo finito $\Gamma$,
con $n := \dim V, \, m := \dim W.$

Sean $g_1,\dots,g_r$ generadores de $\Gamma$, y definamos
$A_i := \rho(g_i) \in \mathrm{GL}(n,\mathbb{F}),\, B_i := \tau(g_i) \in \mathrm{GL}(m,\mathbb{F}), \, i = 1,\dots,r.$

El espacio de homomorfismos $\Gamma$-equivariantes entre $V$ y $W$ es
$\mathrm{Hom}_\Gamma(V,W) = \bigl\{\, X \in \mathrm{Mat}_{m\times n}(\mathbb{F})
     \;:\; X\rho(g) = \tau(g)X \ \ \forall\, g \in \Gamma \,\bigr\}.$
Como $g_1,\dots,g_r$ generan $\Gamma$, esta condición equivale a exigirla
únicamente sobre los generadores:
$$\mathrm{Hom}_\Gamma(V,W) = \bigl\{\, X \in \mathrm{Mat}_{m\times n}(\mathbb{F})
     \;:\; XA_i = B_iX \ \text{para } i=1,\dots,r \,\bigr\}.\tag{$\ast$}$$

## Vectorización

Usamos la identidad estándar del producto de Kronecker:
\begin{align}
\mathrm{vec}(PXQ) = ( P \otimes Q^{T})\,\mathrm{vec}(X) % segun mi convencion de kronecker
\end{align}

Aplicándola a cada lado de $(\ast)$, con $I_m$ e $I_n$ las identidades de
tamaño correspondiente:

\begin{align}
  \mathrm{vec}(XA_i) &= \mathrm{vec}(I_m\,X\,A_i)
    = (I_m \otimes A_i^{T})\,\mathrm{vec}(X), \tag{1}\\[4pt]
  \mathrm{vec}(B_iX) &= \mathrm{vec}(B_i\,X\,I_n)
    = (B_i \otimes I_n)\,\mathrm{vec}(X). \tag{2}
\end{align}

Como $XA_i = B_iX$, los lados derechos de $(1)$ y $(2)$ coinciden:
$$
  (I_m \otimes A_i^{T})\,\mathrm{vec}(X)
  = (B_i \otimes I_n)\,\mathrm{vec}(X),
$$
es decir,
$$
\Bigl[\, (I_m \otimes A_i^{T}) - (B_i \otimes I_n) \,\Bigr]\,\mathrm{vec}(X) = 0,
\qquad i = 1,\dots,r.
\tag{3}
$$

Cada corchete en $(3)$ es una matriz cuadrada de tamaño $mn \times mn$, pues
$\mathrm{vec}(X) \in \mathbb{F}^{mn}$

## Sistema lineal y dimensión

Como $(\ast)$ debe cumplirse simultáneamente para todos los generadores,
apilamos los $r$ bloques de $(3)$ verticalmente en una única matriz:
$$
  M \;=\;
\begin{pmatrix}
    I_m\otimes A_1^{T} \;-\; B_1\otimes I_n \\
    I_m\otimes A_2^{T} \;-\; B_2\otimes I_n \\
\vdots \\
    I_m\otimes A_r^{T} \;-\; B_r\otimes I_n
\end{pmatrix}
  \;\in\; \mathrm{Mat}_{(rmn)\times(mn)}(\mathbb{F}).
$$

Bajo el isomorfismo $X \leftrightarrow \mathrm{vec}(X)$,
$$
\mathrm{Hom}_\Gamma(V,W) \;\cong\; \ker M,
$$
y por el teorema del rango:
$$
\boxed{\;\dim \mathrm{Hom}_\Gamma(V,W) \;=\; mn - \mathrm{rank}(M)\;}
$$

```gap

# ============================================================
# Caso 1: Representación de C2 (un solo generador de orden 2)
# consigo misma. Hom_Gamma(V,V) = End_Gamma(V) = centralizador
# de A en M_2. Como A tiene autovalores distintos (1 y -1),
# el centralizador tiene dimensión 2 (generado por I y A).
# ============================================================
A := [[0,1],[1,0]];;
Print("Caso 1 (esperado 2): ", CppDimHomAModules([A], [A]), "\n");
```

```gap

# ============================================================
# Caso 2: dos representaciones NO isomorfas de C2, ambas 1x1
# (trivial vs. signo). Cualquier X que entrelace debe cumplir
# X*1 = -1*X, o sea X=0. Hom_Gamma(V,W) = 0.
# ============================================================
N := [[1]];;
M := [[-1]];;
Print("Caso 2 (esperado 0): ", CppDimHomAModules([N], [M]), "\n");
```

```gap
# ============================================================
# Caso 3 (corregido): representación regular de C3 consigo
# misma. Por teoría general, dim End_Gamma(FG) = |Gamma|
# siempre (regular rep contra sí misma), sin importar el
# cuerpo ni si el grupo es abeliano.
# ============================================================
g := [[0,0,1],
      [1,0,0],
      [0,1,0]];;

Print("Caso 3 (esperado 3): ",
      CppDimHomAModules([g], [g]), "\n");


# ============================================================
# Caso 3b: Hom de la representación regular de C3 hacia la
# representación trivial (1-dim). X es una fila 1x3 tal que
# X*g = X, o sea X es autovector izquierdo de g con autovalor
# 1. Como g es una matriz de permutación cíclica, ese autoespacio
# es exactamente <(1,1,1)>, dimensión 1. Esto SI ilustra
# la multiplicidad de la componente trivial dentro de la
# representación regular.
# ============================================================
trivial := [[1]];;

Print("Caso 3b (esperado 1): ", CppDimHomAModules([g], [trivial]), "\n");
```

```gap

# ============================================================
# Caso 4: representación trivial de dos generadores (matriz
# identidad para ambos), en dimensión 3 vs dimensión 2.
# Cualquier X de 3x2 entrelaza trivialmente. Hom = 3*2 = 6.
# ============================================================
I3 := IdentityMat(3);;
I2 := IdentityMat(2);;
Print("Caso 4 (esperado 6): ", CppDimHomAModules([I3, I3], [I2, I2]), "\n");
```

```gap
# ============================================================
# Caso 5: chequeo de manejo de errores — cantidad de
# generadores distinta entre Nrep y Mrep. Debe tirar un error
# controlado, no crashear.
# ============================================================
if CppDimHomAModules([A], [A, A]) then
    Print("ERROR: debería haber lanzado una excepción\n");
fi;
```
