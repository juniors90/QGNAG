Only the data stored in each module is used (`simple`, `DeltaStructureMatrices`, `group`, `generatorsofgroup`).

## Block B3. Comparison with the modules of the new text

`paper` are the 56 modules built from the formulas of Sections 2.3.1-2.3.4. `cmp.kind[k]` tells how the basis of `SimplesMn[k]` differs from the text.
Expected: 56/56 isomorphic; identity for 1-12, general for 13-16, permutation for 17-56.

```gap
paper := QGNAG_PaperSimples( G, gens );;
paper_check := QGNAG_CheckSimplesDG( paper );;       # the modules of the text are correct by themselves
cmp := QGNAG_CompareSimplesWithPaper( SimplesMn, paper );;
```
